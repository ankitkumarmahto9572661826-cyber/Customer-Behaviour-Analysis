```python
# ============================================================
#           STEP 5: CHURN ANALYSIS
# ============================================================


# ============================================================
# STEP 1: IMPORT REQUIRED LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully.")


# ============================================================
# STEP 2: LOAD CUSTOMER RFM SEGMENTATION DATA
# ============================================================

# This file was created during Phase 3
rfm = pd.read_csv("customer_rfm_segments.csv")

print("RFM dataset loaded successfully.")

print("\nDataset Shape:")
print(rfm.shape)

print("\nFirst 5 Records:")
display(rfm.head())


# ============================================================
# STEP 3: CHECK COLUMN NAMES
# ============================================================

print("\nAvailable Columns:")
print(rfm.columns.tolist())


# ============================================================
# STEP 4: STANDARDIZE COLUMN NAMES
# ============================================================

# Remove unwanted spaces from column names
rfm.columns = (
    rfm.columns
    .str.strip()
)

print("\nStandardized Columns:")
print(rfm.columns.tolist())


# ============================================================
# STEP 5: CHECK REQUIRED COLUMNS
# ============================================================

required_columns = [
    "customer_id",
    "recency",
    "frequency",
    "monetary",
    "R_Score",
    "F_Score",
    "M_Score",
    "RFM_Score",
    "churn"
]

missing_columns = [
    column
    for column in required_columns
    if column not in rfm.columns
]

if len(missing_columns) > 0:

    print(
        "\nMissing Required Columns:"
    )

    print(missing_columns)

else:

    print(
        "\nAll required columns are available."
    )


# ============================================================
# STEP 6: CHECK DATA TYPES
# ============================================================

print("\nData Types:")
print(rfm.dtypes)


# ============================================================
# STEP 7: CONVERT NUMERICAL COLUMNS
# ============================================================

numeric_columns = [
    "recency",
    "frequency",
    "monetary",
    "R_Score",
    "F_Score",
    "M_Score",
    "RFM_Score",
    "churn"
]

for column in numeric_columns:

    if column in rfm.columns:

        rfm[column] = pd.to_numeric(
            rfm[column],
            errors="coerce"
        )


# ============================================================
# STEP 8: CHECK MISSING VALUES
# ============================================================

print("\nMissing Values:")

missing_values = (
    rfm.isnull()
    .sum()
    .sort_values(
        ascending=False
    )
)

display(
    missing_values[
        missing_values > 0
    ]
)


# ============================================================
# STEP 9: REMOVE INVALID CHURN VALUES
# ============================================================

# Churn should contain only 0 and 1

if "churn" in rfm.columns:

    rfm = rfm[
        rfm["churn"].isin([0, 1])
    ].copy()

print(
    "\nDataset Shape After Churn Validation:",
    rfm.shape
)


# ============================================================
# STEP 10: REMOVE INVALID RFM VALUES
# ============================================================

# RFM values should not be negative

rfm = rfm[
    (rfm["recency"] >= 0) &
    (rfm["frequency"] > 0) &
    (rfm["monetary"] > 0)
].copy()

print(
    "\nDataset Shape After RFM Validation:",
    rfm.shape
)


# ============================================================
# STEP 11: CHECK DUPLICATE CUSTOMERS
# ============================================================

duplicate_customers = (
    rfm["customer_id"]
    .duplicated()
    .sum()
)

print(
    "\nDuplicate Customer IDs:",
    duplicate_customers
)


# Remove duplicate customer records if any
if duplicate_customers > 0:

    rfm = (
        rfm
        .drop_duplicates(
            subset="customer_id",
            keep="first"
        )
        .copy()
    )


# ============================================================
# STEP 12: BASIC RFM STATISTICS
# ============================================================

print("\nRFM Descriptive Statistics:")

display(
    rfm[
        [
            "recency",
            "frequency",
            "monetary",
            "RFM_Score"
        ]
    ].describe()
)


# ============================================================
# STEP 13: OVERALL CHURN COUNT
# ============================================================

churn_count = (
    rfm["churn"]
    .value_counts()
    .sort_index()
)

print("\nChurn Count:")

display(
    churn_count
)


# ============================================================
# STEP 14: OVERALL CHURN RATE
# ============================================================

total_customers = len(rfm)

churned_customers = int(
    rfm["churn"].sum()
)

non_churned_customers = (
    total_customers
    - churned_customers
)

overall_churn_rate = (
    churned_customers
    / total_customers
    * 100
)

print(
    "\nTotal Customers:",
    total_customers
)

print(
    "Churned Customers:",
    churned_customers
)

print(
    "Non-Churned Customers:",
    non_churned_customers
)

print(
    "Overall Churn Rate:",
    round(
        overall_churn_rate,
        2
    ),
    "%"
)


# ============================================================
# STEP 15: CREATE CHURN SUMMARY TABLE
# ============================================================

churn_summary = pd.DataFrame({

    "Churn_Status": [
        "Not Churned",
        "Churned"
    ],

    "Customer_Count": [
        non_churned_customers,
        churned_customers
    ]
})

churn_summary["Percentage"] = (
    churn_summary["Customer_Count"]
    / total_customers
    * 100
)

churn_summary["Percentage"] = (
    churn_summary["Percentage"]
    .round(2)
)

print("\nChurn Summary:")

display(
    churn_summary
)


# ============================================================
# STEP 16: CHURN DISTRIBUTION VISUALIZATION
# ============================================================

plt.figure(
    figsize=(6,3)
)

sns.barplot(
    data=churn_summary,
    x="Churn_Status",
    y="Customer_Count"
)

plt.title(
    "Customer Churn Distribution"
)

plt.xlabel(
    "Churn Status"
)

plt.ylabel(
    "Number of Customers"
)

plt.tight_layout()

plt.show()


# ============================================================
# STEP 17: RFM VALUES BY CHURN STATUS
# ============================================================

rfm_churn_comparison = (
    rfm
    .groupby("churn")
    .agg(

        Customers=(
            "customer_id",
            "count"
        ),

        Average_Recency=(
            "recency",
            "mean"
        ),

        Average_Frequency=(
            "frequency",
            "mean"
        ),

        Average_Monetary=(
            "monetary",
            "mean"
        ),

        Average_RFM_Score=(
            "RFM_Score",
            "mean"
        )
    )
    .reset_index()
)

rfm_churn_comparison[
    "Churn_Status"
] = (
    rfm_churn_comparison[
        "churn"
    ]
    .map({
        0: "Not Churned",
        1: "Churned"
    })
)

rfm_churn_comparison = (
    rfm_churn_comparison
    .round(2)
)

print(
    "\nRFM Comparison by Churn Status:"
)

display(
    rfm_churn_comparison
)


# ============================================================
# STEP 18: RECENCY VS CHURN
# ============================================================

recency_churn = (
    rfm
    .groupby("churn")
    .agg(

        Customers=(
            "customer_id",
            "count"
        ),

        Average_Recency=(
            "recency",
            "mean"
        ),

        Median_Recency=(
            "recency",
            "median"
        )
    )
    .reset_index()
)

recency_churn[
    "Churn_Status"
] = (
    recency_churn[
        "churn"
    ]
    .map({
        0: "Not Churned",
        1: "Churned"
    })
)

print(
    "\nRecency vs Churn:"
)

display(
    recency_churn.round(2)
)


# ------------------------------------------------------------
# RECENCY BOXPLOT
# ------------------------------------------------------------

plt.figure(
    figsize=(6,3)
)

sns.boxplot(
    data=rfm,
    x="churn",
    y="recency"
)

plt.title(
    "Recency by Churn Status"
)

plt.xlabel(
    "Churn Status"
)

plt.ylabel(
    "Recency (Days)"
)

plt.xticks(
    [0, 1],
    [
        "Not Churned",
        "Churned"
    ]
)

plt.tight_layout()

plt.show()


# ============================================================
# STEP 19: FREQUENCY VS CHURN
# ============================================================

frequency_churn = (
    rfm
    .groupby("churn")
    .agg(

        Customers=(
            "customer_id",
            "count"
        ),

        Average_Frequency=(
            "frequency",
            "mean"
        ),

        Median_Frequency=(
            "frequency",
            "median"
        )
    )
    .reset_index()
)

frequency_churn[
    "Churn_Status"
] = (
    frequency_churn[
        "churn"
    ]
    .map({
        0: "Not Churned",
        1: "Churned"
    })
)

print(
    "\nFrequency vs Churn:"
)

display(
    frequency_churn.round(2)
)


# ------------------------------------------------------------
# FREQUENCY BOXPLOT
# ------------------------------------------------------------

plt.figure(
    figsize=(6,3)
)

sns.boxplot(
    data=rfm,
    x="churn",
    y="frequency"
)

plt.title(
    "Purchase Frequency by Churn Status"
)

plt.xlabel(
    "Churn Status"
)

plt.ylabel(
    "Purchase Frequency"
)

plt.xticks(
    [0, 1],
    [
        "Not Churned",
        "Churned"
    ]
)

plt.tight_layout()

plt.show()


# ============================================================
# STEP 20: MONETARY VALUE VS CHURN
# ============================================================

monetary_churn = (
    rfm
    .groupby("churn")
    .agg(

        Customers=(
            "customer_id",
            "count"
        ),

        Average_Spending=(
            "monetary",
            "mean"
        ),

        Median_Spending=(
            "monetary",
            "median"
        )
    )
    .reset_index()
)

monetary_churn[
    "Churn_Status"
] = (
    monetary_churn[
        "churn"
    ]
    .map({
        0: "Not Churned",
        1: "Churned"
    })
)

print(
    "\nMonetary Value vs Churn:"
)

display(
    monetary_churn.round(2)
)


# ------------------------------------------------------------
# MONETARY BOXPLOT
# ------------------------------------------------------------

plt.figure(
    figsize=(6,3)
)

sns.boxplot(
    data=rfm,
    x="churn",
    y="monetary"
)

plt.title(
    "Customer Spending by Churn Status"
)

plt.xlabel(
    "Churn Status"
)

plt.ylabel(
    "Total Spending"
)

plt.xticks(
    [0, 1],
    [
        "Not Churned",
        "Churned"
    ]
)

plt.tight_layout()

plt.show()


# ============================================================
# STEP 21: RFM SCORE VS CHURN
# ============================================================

rfm_score_churn = (
    rfm
    .groupby("churn")
    .agg(

        Customers=(
            "customer_id",
            "count"
        ),

        Average_RFM_Score=(
            "RFM_Score",
            "mean"
        ),

        Median_RFM_Score=(
            "RFM_Score",
            "median"
        )
    )
    .reset_index()
)

rfm_score_churn[
    "Churn_Status"
] = (
    rfm_score_churn[
        "churn"
    ]
    .map({
        0: "Not Churned",
        1: "Churned"
    })
)

print(
    "\nRFM Score vs Churn:"
)

display(
    rfm_score_churn.round(2)
)


# ------------------------------------------------------------
# RFM SCORE BOXPLOT
# ------------------------------------------------------------

plt.figure(
    figsize=(6,3)
)

sns.boxplot(
    data=rfm,
    x="churn",
    y="RFM_Score"
)

plt.title(
    "RFM Score by Churn Status"
)

plt.xlabel(
    "Churn Status"
)

plt.ylabel(
    "RFM Score"
)

plt.xticks(
    [0, 1],
    [
        "Not Churned",
        "Churned"
    ]
)

plt.tight_layout()

plt.show()


# ============================================================
# STEP 22: CHURN RATE BY RFM SCORE
# ============================================================

churn_by_rfm_score = (
    rfm
    .groupby("RFM_Score")
    .agg(

        Customers=(
            "customer_id",
            "count"
        ),

        Churned_Customers=(
            "churn",
            "sum"
        ),

        Churn_Rate=(
            "churn",
            "mean"
        )
    )
    .reset_index()
)

churn_by_rfm_score[
    "Churn_Rate"
] = (
    churn_by_rfm_score[
        "Churn_Rate"
    ] * 100
)

churn_by_rfm_score = (
    churn_by_rfm_score
    .round(2)
)

print(
    "\nChurn Rate by RFM Score:"
)

display(
    churn_by_rfm_score
)


# ------------------------------------------------------------
# RFM SCORE CHURN TREND
# ------------------------------------------------------------

plt.figure(
    figsize=(6,3)
)

sns.lineplot(
    data=churn_by_rfm_score,
    x="RFM_Score",
    y="Churn_Rate",
    marker="o"
)

plt.title(
    "Churn Rate by RFM Score"
)

plt.xlabel(
    "RFM Score"
)

plt.ylabel(
    "Churn Rate (%)"
)

plt.tight_layout()

plt.show()


# ============================================================
# STEP 23: CHURN BY CUSTOMER SEGMENT
# ============================================================

if "Customer_Segment" in rfm.columns:

    segment_churn = (
        rfm
        .groupby("Customer_Segment")
        .agg(

            Customers=(
                "customer_id",
                "count"
            ),

            Churned_Customers=(
                "churn",
                "sum"
            ),

            Churn_Rate=(
                "churn",
                "mean"
            ),

            Average_Spending=(
                "monetary",
                "mean"
            ),

            Average_Frequency=(
                "frequency",
                "mean"
            ),

            Average_Recency=(
                "recency",
                "mean"
            )
        )
        .reset_index()
    )

    segment_churn[
        "Churn_Rate"
    ] = (
        segment_churn[
            "Churn_Rate"
        ] * 100
    )

    segment_churn = (
        segment_churn
        .round(2)
        .sort_values(
            "Churn_Rate",
            ascending=False
        )
    )

    print(
        "\nChurn by Customer Segment:"
    )

    display(
        segment_churn
    )

else:

    print(
        "\nCustomer_Segment column not available."
    )

    segment_churn = pd.DataFrame()


# ------------------------------------------------------------
# SEGMENT CHURN VISUALIZATION
# ------------------------------------------------------------

if not segment_churn.empty:

    plt.figure(
        figsize=(6,3)
    )

    sns.barplot(
        data=segment_churn,
        x="Churn_Rate",
        y="Customer_Segment"
    )

    plt.title(
        "Churn Rate by Customer Segment"
    )

    plt.xlabel(
        "Churn Rate (%)"
    )

    plt.ylabel(
        "Customer Segment"
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# STEP 24: CHURN BY RECENCY GROUP
# ============================================================

# Create recency groups

rfm["Recency_Group"] = pd.cut(

    rfm["recency"],

    bins=[
        -1,
        30,
        60,
        90,
        180,
        365,
        np.inf
    ],

    labels=[
        "0-30 Days",
        "31-60 Days",
        "61-90 Days",
        "91-180 Days",
        "181-365 Days",
        "365+ Days"
    ]
)


# Calculate churn

recency_group_churn = (
    rfm
    .groupby(
        "Recency_Group",
        observed=True
    )
    .agg(

        Customers=(
            "customer_id",
            "count"
        ),

        Churned_Customers=(
            "churn",
            "sum"
        ),

        Churn_Rate=(
            "churn",
            "mean"
        )
    )
    .reset_index()
)

recency_group_churn[
    "Churn_Rate"
] = (
    recency_group_churn[
        "Churn_Rate"
    ] * 100
)

recency_group_churn = (
    recency_group_churn
    .round(2)
)

print(
    "\nChurn by Recency Group:"
)

display(
    recency_group_churn
)


# ------------------------------------------------------------
# RECENCY GROUP VISUALIZATION
# ------------------------------------------------------------

plt.figure(
    figsize=(6,3)
)

sns.barplot(
    data=recency_group_churn,
    x="Recency_Group",
    y="Churn_Rate"
)

plt.title(
    "Churn Rate by Recency Group"
)

plt.xlabel(
    "Recency Group"
)

plt.ylabel(
    "Churn Rate (%)"
)

plt.xticks(
    rotation=30
)

plt.tight_layout()

plt.show()


# ============================================================
# STEP 25: CHURN BY FREQUENCY GROUP
# ============================================================

# Create frequency groups

rfm["Frequency_Group"] = pd.cut(

    rfm["frequency"],

    bins=[
        0,
        1,
        2,
        5,
        10,
        np.inf
    ],

    labels=[
        "1 Purchase",
        "2 Purchases",
        "3-5 Purchases",
        "6-10 Purchases",
        "10+ Purchases"
    ]
)


# Calculate churn

frequency_group_churn = (
    rfm
    .groupby(
        "Frequency_Group",
        observed=True
    )
    .agg(

        Customers=(
            "customer_id",
            "count"
        ),

        Churned_Customers=(
            "churn",
            "sum"
        ),

        Churn_Rate=(
            "churn",
            "mean"
        )
    )
    .reset_index()
)

frequency_group_churn[
    "Churn_Rate"
] = (
    frequency_group_churn[
        "Churn_Rate"
    ] * 100
)

frequency_group_churn = (
    frequency_group_churn
    .round(2)
)

print(
    "\nChurn by Frequency Group:"
)

display(
    frequency_group_churn
)


# ------------------------------------------------------------
# FREQUENCY GROUP VISUALIZATION
# ------------------------------------------------------------

plt.figure(
    figsize=(6,3)
)

sns.barplot(
    data=frequency_group_churn,
    x="Frequency_Group",
    y="Churn_Rate"
)

plt.title(
    "Churn Rate by Purchase Frequency"
)

plt.xlabel(
    "Purchase Frequency Group"
)

plt.ylabel(
    "Churn Rate (%)"
)

plt.xticks(
    rotation=30
)

plt.tight_layout()

plt.show()


# ============================================================
# STEP 26: CHURN BY MONETARY GROUP
# ============================================================

# Create five spending groups using ranking.
# Ranking prevents qcut errors caused by tied values.

rfm["Monetary_Group"] = pd.qcut(

    rfm["monetary"].rank(
        method="first"
    ),

    q=5,

    labels=[
        "Very Low",
        "Low",
        "Medium",
        "High",
        "Very High"
    ]
)


# Calculate churn

monetary_group_churn = (
    rfm
    .groupby(
        "Monetary_Group",
        observed=True
    )
    .agg(

        Customers=(
            "customer_id",
            "count"
        ),

        Churned_Customers=(
            "churn",
            "sum"
        ),

        Churn_Rate=(
            "churn",
            "mean"
        )
    )
    .reset_index()
)

monetary_group_churn[
    "Churn_Rate"
] = (
    monetary_group_churn[
        "Churn_Rate"
    ] * 100
)

monetary_group_churn = (
    monetary_group_churn
    .round(2)
)

print(
    "\nChurn by Monetary Group:"
)

display(
    monetary_group_churn
)


# ------------------------------------------------------------
# MONETARY GROUP VISUALIZATION
# ------------------------------------------------------------

plt.figure(
    figsize=(6,3)
)

sns.barplot(
    data=monetary_group_churn,
    x="Monetary_Group",
    y="Churn_Rate"
)

plt.title(
    "Churn Rate by Customer Spending Group"
)

plt.xlabel(
    "Spending Group"
)

plt.ylabel(
    "Churn Rate (%)"
)

plt.tight_layout()

plt.show()


# ============================================================
# STEP 27: CHURN BY AGE GROUP
# ============================================================

if "customer_age" in rfm.columns:

    # Create age groups

    rfm["Age_Group"] = pd.cut(

        rfm["customer_age"],

        bins=[
            0,
            18,
            25,
            35,
            45,
            55,
            65,
            100
        ],

        labels=[
            "Under 18",
            "18-25",
            "26-35",
            "36-45",
            "46-55",
            "56-65",
            "65+"
        ],

        include_lowest=True
    )


    # Calculate churn

    age_churn = (
        rfm
        .groupby(
            "Age_Group",
            observed=True
        )
        .agg(

            Customers=(
                "customer_id",
                "count"
            ),

            Churned_Customers=(
                "churn",
                "sum"
            ),

            Churn_Rate=(
                "churn",
                "mean"
            )
        )
        .reset_index()
    )


    # Convert to percentage

    age_churn[
        "Churn_Rate"
    ] = (
        age_churn[
            "Churn_Rate"
        ] * 100
    )


    age_churn = (
        age_churn
        .round(2)
    )


    print(
        "\nChurn by Age Group:"
    )

    display(
        age_churn
    )


    # Visualization

    plt.figure(
        figsize=(6,3)
    )

    sns.barplot(
        data=age_churn,
        x="Age_Group",
        y="Churn_Rate"
    )

    plt.title(
        "Churn Rate by Age Group"
    )

    plt.xlabel(
        "Age Group"
    )

    plt.ylabel(
        "Churn Rate (%)"
    )

    plt.xticks(
        rotation=30
    )

    plt.tight_layout()

    plt.show()

else:

    print(
        "\ncustomer_age column is not available."
    )

    age_churn = pd.DataFrame()


# ============================================================
# STEP 28: CHURN BY GENDER
# ============================================================

if "gender" in rfm.columns:

    gender_churn = (
        rfm
        .groupby("gender")
        .agg(

            Customers=(
                "customer_id",
                "count"
            ),

            Churned_Customers=(
                "churn",
                "sum"
            ),

            Churn_Rate=(
                "churn",
                "mean"
            )
        )
        .reset_index()
    )


    # Convert churn rate to percentage

    gender_churn[
        "Churn_Rate"
    ] = (
        gender_churn[
            "Churn_Rate"
        ] * 100
    )


    gender_churn = (
        gender_churn
        .round(2)
    )


    print(
        "\nChurn by Gender:"
    )

    display(
        gender_churn
    )


    # Visualization

    plt.figure(
        figsize=(6,3)
    )

    sns.barplot(
        data=gender_churn,
        x="gender",
        y="Churn_Rate"
    )

    plt.title(
        "Churn Rate by Gender"
    )

    plt.xlabel(
        "Gender"
    )

    plt.ylabel(
        "Churn Rate (%)"
    )

    plt.tight_layout()

    plt.show()

else:

    print(
        "\ngender column is not available."
    )

    gender_churn = pd.DataFrame()


# ============================================================
# STEP 29: RFM SCORE COMPONENT ANALYSIS
# ============================================================


# ------------------------------------------------------------
# RECENCY SCORE VS CHURN
# ------------------------------------------------------------

r_score_churn = (
    rfm
    .groupby("R_Score")
    .agg(

        Customers=(
            "customer_id",
            "count"
        ),

        Churned_Customers=(
            "churn",
            "sum"
        ),

        Churn_Rate=(
            "churn",
            "mean"
        )
    )
    .reset_index()
)

r_score_churn[
    "Churn_Rate"
] *= 100

r_score_churn = (
    r_score_churn
    .round(2)
)

print(
    "\nChurn by Recency Score:"
)

display(
    r_score_churn
)


# ------------------------------------------------------------
# FREQUENCY SCORE VS CHURN
# ------------------------------------------------------------

f_score_churn = (
    rfm
    .groupby("F_Score")
    .agg(

        Customers=(
            "customer_id",
            "count"
        ),

        Churned_Customers=(
            "churn",
            "sum"
        ),

        Churn_Rate=(
            "churn",
            "mean"
        )
    )
    .reset_index()
)

f_score_churn[
    "Churn_Rate"
] *= 100

f_score_churn = (
    f_score_churn
    .round(2)
)

print(
    "\nChurn by Frequency Score:"
)

display(
    f_score_churn
)


# ------------------------------------------------------------
# MONETARY SCORE VS CHURN
# ------------------------------------------------------------

m_score_churn = (
    rfm
    .groupby("M_Score")
    .agg(

        Customers=(
            "customer_id",
            "count"
        ),

        Churned_Customers=(
            "churn",
            "sum"
        ),

        Churn_Rate=(
            "churn",
            "mean"
        )
    )
    .reset_index()
)

m_score_churn[
    "Churn_Rate"
] *= 100

m_score_churn = (
    m_score_churn
    .round(2)
)

print(
    "\nChurn by Monetary Score:"
)

display(
    m_score_churn
)


# ============================================================
# STEP 30: RFM CHURN HEATMAP
# ============================================================

rfm_churn_heatmap = (
    rfm
    .pivot_table(
        index="R_Score",
        columns="F_Score",
        values="churn",
        aggfunc="mean"
    )
)

# Convert to percentage

rfm_churn_heatmap = (
    rfm_churn_heatmap * 100
)

print(
    "\nRFM Churn Heatmap Data:"
)

display(
    rfm_churn_heatmap.round(2)
)


plt.figure(
    figsize=(6,3)
)

sns.heatmap(
    rfm_churn_heatmap,
    annot=True,
    fmt=".1f"
)

plt.title(
    "Churn Rate (%) by Recency Score and Frequency Score"
)

plt.xlabel(
    "Frequency Score"
)

plt.ylabel(
    "Recency Score"
)

plt.tight_layout()

plt.show()


# ============================================================
# STEP 31: IDENTIFY HIGH-RISK CUSTOMERS
# ============================================================

# High-risk conditions:
#
# 1. Already churned
# OR
# 2. Very low recency score
# OR
# 3. Very low overall RFM score

high_risk_customers = rfm[
    (
        (rfm["churn"] == 1)
        |
        (rfm["R_Score"] <= 2)
        |
        (rfm["RFM_Score"] <= 6)
    )
].copy()


print(
    "\nHigh-Risk Customers:",
    len(high_risk_customers)
)


# Display high-risk customers

display(
    high_risk_customers[
        [
            "customer_id",
            "recency",
            "frequency",
            "monetary",
            "R_Score",
            "F_Score",
            "M_Score",
            "RFM_Score",
            "churn"
        ]
    ]
    .sort_values(
        "monetary",
        ascending=False
    )
    .head(20)
)


# ============================================================
# STEP 32: HIGH-VALUE AT-RISK CUSTOMERS
# ============================================================

# High-value customers with poor recency

high_value_at_risk = rfm[
    (
        (rfm["R_Score"] <= 2)
        &
        (rfm["M_Score"] >= 4)
        &
        (rfm["churn"] == 0)
    )
].copy()


print(
    "\nHigh-Value At-Risk Customers:",
    len(high_value_at_risk)
)


display(
    high_value_at_risk[
        [
            "customer_id",
            "recency",
            "frequency",
            "monetary",
            "R_Score",
            "F_Score",
            "M_Score",
            "RFM_Score",
            "churn"
        ]
    ]
    .sort_values(
        "monetary",
        ascending=False
    )
    .head(20)
)


# ============================================================
# STEP 33: CHURNED HIGH-VALUE CUSTOMERS
# ============================================================

# Customers who have churned
# and belong to the high monetary-value group

churned_high_value = rfm[
    (
        (rfm["churn"] == 1)
        &
        (rfm["M_Score"] >= 4)
    )
].copy()


print(
    "\nChurned High-Value Customers:",
    len(churned_high_value)
)


display(
    churned_high_value[
        [
            "customer_id",
            "recency",
            "frequency",
            "monetary",
            "R_Score",
            "F_Score",
            "M_Score",
            "RFM_Score"
        ]
    ]
    .sort_values(
        "monetary",
        ascending=False
    )
    .head(20)
)


# ============================================================
# STEP 34: REVENUE IMPACT OF CHURN
# ============================================================

# Revenue associated with churned customers

churned_revenue = (
    rfm
    .loc[
        rfm["churn"] == 1,
        "monetary"
    ]
    .sum()
)


# Revenue associated with active customers

active_revenue = (
    rfm
    .loc[
        rfm["churn"] == 0,
        "monetary"
    ]
    .sum()
)


# Total revenue

total_customer_revenue = (
    rfm["monetary"]
    .sum()
)


# Calculate percentage

if total_customer_revenue > 0:

    churned_revenue_percentage = (
        churned_revenue
        / total_customer_revenue
        * 100
    )

else:

    churned_revenue_percentage = 0


print(
    "\nChurned Customer Revenue:",
    round(
        churned_revenue,
        2
    )
)

print(
    "Active Customer Revenue:",
    round(
        active_revenue,
        2
    )
)

print(
    "Total Customer Revenue:",
    round(
        total_customer_revenue,
        2
    )
)

print(
    "Churned Revenue Percentage:",
    round(
        churned_revenue_percentage,
        2
    ),
    "%"
)


# ============================================================
# STEP 35: CREATE CHURN RISK CATEGORIES
# ============================================================

def assign_churn_risk(row):

    # Customers who are already churned
    if row["churn"] == 1:

        return "Churned"

    # High-risk customers
    elif (
        row["R_Score"] <= 2
        and
        row["F_Score"] <= 2
    ):

        return "High Risk"

    # Medium-risk customers
    elif row["R_Score"] <= 2:

        return "Medium Risk"

    # Low-risk customers
    elif row["R_Score"] >= 4:

        return "Low Risk"

    # Remaining customers
    else:

        return "Moderate Risk"


# Apply function

rfm["Churn_Risk"] = (
    rfm.apply(
        assign_churn_risk,
        axis=1
    )
)


# ============================================================
# STEP 36: CHURN RISK SUMMARY
# ============================================================

risk_summary = (
    rfm["Churn_Risk"]
    .value_counts()
    .reset_index()
)

risk_summary.columns = [
    "Churn_Risk",
    "Customer_Count"
]

risk_summary["Percentage"] = (
    risk_summary["Customer_Count"]
    / len(rfm)
    * 100
)

risk_summary["Percentage"] = (
    risk_summary["Percentage"]
    .round(2)
)


print(
    "\nChurn Risk Summary:"
)

display(
    risk_summary
)


# ============================================================
# STEP 37: CHURN RISK VISUALIZATION
# ============================================================

plt.figure(
    figsize=(6,3)
)

sns.barplot(
    data=risk_summary,
    x="Customer_Count",
    y="Churn_Risk"
)

plt.title(
    "Customer Churn Risk Distribution"
)

plt.xlabel(
    "Number of Customers"
)

plt.ylabel(
    "Churn Risk"
)

plt.tight_layout()

plt.show()


# ============================================================
# STEP 38: CORRELATION ANALYSIS
# ============================================================

correlation_columns = [
    "recency",
    "frequency",
    "monetary",
    "R_Score",
    "F_Score",
    "M_Score",
    "RFM_Score",
    "churn"
]


# Keep only available columns

available_columns = [
    column
    for column in correlation_columns
    if column in rfm.columns
]


# Create correlation matrix

correlation_matrix = (
    rfm[
        available_columns
    ]
    .corr()
)


print(
    "\nCorrelation Matrix:"
)

display(
    correlation_matrix.round(2)
)


# ============================================================
# STEP 39: CORRELATION HEATMAP
# ============================================================

plt.figure(
    figsize=(6,3)
)

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f"
)

plt.title(
    "RFM and Churn Correlation Matrix"
)

plt.tight_layout()

plt.show()


# ============================================================
# STEP 40: FINAL CHURN SUMMARY
# ============================================================

final_churn_summary = pd.DataFrame({

    "Metric": [

        "Total Customers",

        "Churned Customers",

        "Non-Churned Customers",

        "Overall Churn Rate (%)",

        "Average Recency - Churned",

        "Average Recency - Non-Churned",

        "Average Frequency - Churned",

        "Average Frequency - Non-Churned",

        "Average Spending - Churned",

        "Average Spending - Non-Churned",

        "Churned Customer Revenue",

        "Churned Revenue Percentage (%)",

        "High-Risk Customers",

        "High-Value At-Risk Customers",

        "Churned High-Value Customers"
    ],

    "Value": [

        total_customers,

        churned_customers,

        non_churned_customers,

        round(
            overall_churn_rate,
            2
        ),

        round(
            rfm.loc[
                rfm["churn"] == 1,
                "recency"
            ].mean(),
            2
        ),

        round(
            rfm.loc[
                rfm["churn"] == 0,
                "recency"
            ].mean(),
            2
        ),

        round(
            rfm.loc[
                rfm["churn"] == 1,
                "frequency"
            ].mean(),
            2
        ),

        round(
            rfm.loc[
                rfm["churn"] == 0,
                "frequency"
            ].mean(),
            2
        ),

        round(
            rfm.loc[
                rfm["churn"] == 1,
                "monetary"
            ].mean(),
            2
        ),

        round(
            rfm.loc[
                rfm["churn"] == 0,
                "monetary"
            ].mean(),
            2
        ),

        round(
            churned_revenue,
            2
        ),

        round(
            churned_revenue_percentage,
            2
        ),

        len(
            high_risk_customers
        ),

        len(
            high_value_at_risk
        ),

        len(
            churned_high_value
        )
    ]
})


print(
    "\nFinal Churn Analysis Summary:"
)

display(
    final_churn_summary
)


# ============================================================
# STEP 41: SAVE COMPLETE CHURN DATASET
# ============================================================

rfm.to_csv(
    "customer_churn_analysis.csv",
    index=False
)

print(
    "Saved: customer_churn_analysis.csv"
)


# ============================================================
# STEP 42: SAVE CHURN SUMMARY
# ============================================================

churn_summary.to_csv(
    "overall_churn_summary.csv",
    index=False
)

final_churn_summary.to_csv(
    "final_churn_summary.csv",
    index=False
)

print(
    "Saved: overall_churn_summary.csv"
)

print(
    "Saved: final_churn_summary.csv"
)


# ============================================================
# STEP 43: SAVE RFM CHURN ANALYSIS
# ============================================================

rfm_churn_comparison.to_csv(
    "rfm_churn_comparison.csv",
    index=False
)

churn_by_rfm_score.to_csv(
    "rfm_score_churn_analysis.csv",
    index=False
)

print(
    "Saved: RFM churn analysis files."
)


# ============================================================
# STEP 44: SAVE GROUP ANALYSIS FILES
# ============================================================

recency_group_churn.to_csv(
    "recency_group_churn.csv",
    index=False
)

frequency_group_churn.to_csv(
    "frequency_group_churn.csv",
    index=False
)

monetary_group_churn.to_csv(
    "monetary_group_churn.csv",
    index=False
)

print(
    "Saved: Recency, Frequency and Monetary group files."
)


# ============================================================
# STEP 45: SAVE AGE AND GENDER ANALYSIS
# ============================================================

if not age_churn.empty:

    age_churn.to_csv(
        "age_group_churn.csv",
        index=False
    )

    print(
        "Saved: age_group_churn.csv"
    )


if not gender_churn.empty:

    gender_churn.to_csv(
        "gender_churn_analysis.csv",
        index=False
    )

    print(
        "Saved: gender_churn_analysis.csv"
    )


# ============================================================
# STEP 46: SAVE RFM SCORE COMPONENT ANALYSIS
# ============================================================

r_score_churn.to_csv(
    "recency_score_churn.csv",
    index=False
)

f_score_churn.to_csv(
    "frequency_score_churn.csv",
    index=False
)

m_score_churn.to_csv(
    "monetary_score_churn.csv",
    index=False
)

print(
    "Saved: RFM score component files."
)


# ============================================================
# STEP 47: SAVE RFM HEATMAP DATA
# ============================================================

rfm_churn_heatmap.to_csv(
    "rfm_churn_heatmap.csv"
)

print(
    "Saved: rfm_churn_heatmap.csv"
)


# ============================================================
# STEP 48: SAVE SEGMENT CHURN ANALYSIS
# ============================================================

if not segment_churn.empty:

    segment_churn.to_csv(
        "segment_churn_analysis.csv",
        index=False
    )

    print(
        "Saved: segment_churn_analysis.csv"
    )


# ============================================================
# STEP 49: SAVE HIGH-RISK CUSTOMER FILES
# ============================================================

high_risk_customers.to_csv(
    "high_risk_customers.csv",
    index=False
)

high_value_at_risk.to_csv(
    "high_value_at_risk_customers.csv",
    index=False
)

churned_high_value.to_csv(
    "churned_high_value_customers.csv",
    index=False
)

print(
    "Saved: High-risk customer files."
)


# ============================================================
# STEP 50: FINAL STEP 5 REPORT
# ============================================================

print("\n")
print("=" * 70)
print("STEP 5: CHURN ANALYSIS COMPLETED SUCCESSFULLY")
print("=" * 70)

print(
    "\nTotal Customers:",
    total_customers
)

print(
    "Churned Customers:",
    churned_customers
)

print(
    "Non-Churned Customers:",
    non_churned_customers
)

print(
    "Overall Churn Rate:",
    round(
        overall_churn_rate,
        2
    ),
    "%"
)

print(
    "High-Risk Customers:",
    len(high_risk_customers)
)

print(
    "High-Value At-Risk Customers:",
    len(high_value_at_risk)
)

print(
    "Churned High-Value Customers:",
    len(churned_high_value)
)

print(
    "Churned Customer Revenue:",
    round(
        churned_revenue,
        2
    )
)


print("\n")
print("=" * 70)
print("STEP 5 OUTPUT FILES")
print("=" * 70)

print("1. customer_churn_analysis.csv")
print("2. overall_churn_summary.csv")
print("3. final_churn_summary.csv")
print("4. rfm_churn_comparison.csv")
print("5. rfm_score_churn_analysis.csv")
print("6. recency_group_churn.csv")
print("7. frequency_group_churn.csv")
print("8. monetary_group_churn.csv")
print("9. age_group_churn.csv")
print("10. gender_churn_analysis.csv")
print("11. recency_score_churn.csv")
print("12. frequency_score_churn.csv")
print("13. monetary_score_churn.csv")
print("14. rfm_churn_heatmap.csv")
print("15. segment_churn_analysis.csv")
print("16. high_risk_customers.csv")
print("17. high_value_at_risk_customers.csv")
print("18. churned_high_value_customers.csv")

print("\nStep 5 completed.") 
```
