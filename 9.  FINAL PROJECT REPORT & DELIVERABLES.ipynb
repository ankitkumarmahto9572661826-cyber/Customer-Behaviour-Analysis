```python
# ============================================================
#     STEP 9 — FINAL PROJECT REPORT & DELIVERABLES
# ============================================================
# Goal:
# Create the final project deliverables:
#
# 1. Final project summary
# 2. Business KPI report
# 3. Customer insights report
# 4. Product insights report
# 5. Purchase behavior report
# 6. Churn report
# 7. RFM segmentation report
# 8. Exactly 5 business recommendations
# 9. Project README
# 10. Final submission folder
# 11. ZIP file containing the complete project
#
# ============================================================


# ------------------------------------------------------------
# STEP 1 — IMPORT LIBRARIES
# ------------------------------------------------------------

import pandas as pd
import numpy as np
from pathlib import Path
import shutil
from datetime import datetime


# ============================================================
# STEP 2 — DEFINE PROJECT FOLDERS
# ============================================================

project_folder = Path("Customer_Behavior_Analysis_Final")

project_folder.mkdir(
    exist_ok=True
)

data_folder = project_folder / "Data"
reports_folder = project_folder / "Reports"
powerbi_folder = project_folder / "PowerBI"
insights_folder = project_folder / "Business_Insights"

data_folder.mkdir(exist_ok=True)
reports_folder.mkdir(exist_ok=True)
powerbi_folder.mkdir(exist_ok=True)
insights_folder.mkdir(exist_ok=True)


# ============================================================
# STEP 3 — LOAD FINAL DATASETS
# ============================================================

transactions = pd.read_csv(
    "ecommerce_cleaned_transactions.csv"
)

customers = pd.read_csv(
    "customer_level_features.csv"
)

rfm = pd.read_csv(
    "customer_rfm_segments.csv"
)


# ------------------------------------------------------------
# LOAD STEP 7 FILES
# ------------------------------------------------------------

step7_folder = Path(
    "step7_powerbi_data"
)

fact_sales = pd.read_csv(
    step7_folder / "fact_sales.csv"
)

customer_kpis = pd.read_csv(
    step7_folder / "customer_kpis.csv"
)

monthly_sales = pd.read_csv(
    step7_folder / "monthly_sales.csv"
)

category_performance = pd.read_csv(
    step7_folder / "category_performance.csv"
)

payment_performance = pd.read_csv(
    step7_folder / "payment_performance.csv"
)

day_performance = pd.read_csv(
    step7_folder / "day_performance.csv"
)

hour_performance = pd.read_csv(
    step7_folder / "hour_performance.csv"
)

time_period_performance = pd.read_csv(
    step7_folder / "time_period_performance.csv"
)


# ============================================================
# STEP 4 — LOAD STEP 8 INSIGHTS
# ============================================================

step8_folder = Path(
    "step8_business_insights"
)

business_kpis = pd.read_csv(
    step8_folder / "business_kpis.csv"
)

product_insights = pd.read_csv(
    step8_folder / "product_insights.csv"
)

purchase_insights = pd.read_csv(
    step8_folder / "purchase_behavior_insights.csv"
)

customer_insights = pd.read_csv(
    step8_folder / "customer_value_insights.csv"
)

rfm_insights = pd.read_csv(
    step8_folder / "rfm_insights.csv"
)

churn_insights = pd.read_csv(
    step8_folder / "churn_insights.csv"
)

payment_insights = pd.read_csv(
    step8_folder / "payment_insights.csv"
)

recommendations = pd.read_csv(
    step8_folder / "business_recommendations.csv"
)


# ============================================================
# STEP 5 — COPY IMPORTANT DATASETS
# ============================================================

# ------------------------------------------------------------
# Copy cleaned transaction data
# ------------------------------------------------------------

shutil.copy2(
    "ecommerce_cleaned_transactions.csv",
    data_folder / "ecommerce_cleaned_transactions.csv"
)


# ------------------------------------------------------------
# Copy customer-level data
# ------------------------------------------------------------

shutil.copy2(
    "customer_level_features.csv",
    data_folder / "customer_level_features.csv"
)


# ------------------------------------------------------------
# Copy RFM data
# ------------------------------------------------------------

shutil.copy2(
    "customer_rfm_segments.csv",
    data_folder / "customer_rfm_segments.csv"
)


# ============================================================
# STEP 6 — COPY POWER BI DATASETS
# ============================================================

for file in step7_folder.glob("*.csv"):

    shutil.copy2(
        file,
        powerbi_folder / file.name
    )


# ============================================================
# STEP 7 — CREATE FINAL KPI TABLE
# ============================================================

total_customers = fact_sales[
    "customer_id"
].nunique()

total_transactions = len(
    fact_sales
)

total_revenue = fact_sales[
    "total_purchase_amount"
].sum()

average_order_value = (
    fact_sales[
        "total_purchase_amount"
    ].mean()
)

total_quantity = fact_sales[
    "quantity"
].sum()

if "Return_Flag" in fact_sales.columns:

    total_returns = (
        fact_sales["Return_Flag"].sum()
    )

    return_rate = (
        total_returns /
        total_transactions
    ) * 100

else:

    total_returns = 0
    return_rate = 0


if "churn" in customer_kpis.columns:

    churned_customers = (
        customer_kpis["churn"]
        .fillna(0)
        .sum()
    )

    churn_rate = (
        churned_customers /
        total_customers
    ) * 100

else:

    churned_customers = 0
    churn_rate = 0


final_kpis = pd.DataFrame({
    "KPI": [
        "Total Customers",
        "Total Transactions",
        "Total Revenue",
        "Average Order Value",
        "Total Quantity Sold",
        "Total Returns",
        "Return Rate (%)",
        "Churned Customers",
        "Churn Rate (%)"
    ],
    "Value": [
        total_customers,
        total_transactions,
        total_revenue,
        average_order_value,
        total_quantity,
        total_returns,
        return_rate,
        churned_customers,
        churn_rate
    ]
})


# ============================================================
# STEP 8 — SAVE FINAL KPI REPORT
# ============================================================

final_kpis.to_csv(
    reports_folder / "Final_KPI_Report.csv",
    index=False
)


# ============================================================
# STEP 9 — IDENTIFY KEY BUSINESS FINDINGS
# ============================================================

best_category = (
    category_performance
    .loc[
        category_performance["Revenue"].idxmax(),
        "product_category"
    ]
)

best_category_revenue = (
    category_performance["Revenue"].max()
)


most_used_payment = (
    payment_performance
    .loc[
        payment_performance["Transactions"].idxmax(),
        "payment_method"
    ]
)


best_purchase_day = (
    day_performance
    .loc[
        day_performance["Transactions"].idxmax(),
        "Day_Name"
    ]
)


best_purchase_hour = (
    hour_performance
    .loc[
        hour_performance["Transactions"].idxmax(),
        "Purchase_Hour"
    ]
)


best_month = (
    monthly_sales
    .loc[
        monthly_sales["Revenue"].idxmax(),
        "Month_Year"
    ]
)


best_month_revenue = (
    monthly_sales["Revenue"].max()
)


# ============================================================
# STEP 10 — CUSTOMER SEGMENT FINDINGS
# ============================================================

if "Customer_Segment" in rfm.columns:

    segment_summary = (
        rfm
        .groupby(
            "Customer_Segment",
            as_index=False
        )
        .agg(
            Customers=(
                "customer_id",
                "nunique"
            ),
            Revenue=(
                "monetary",
                "sum"
            ),
            Average_Spend=(
                "monetary",
                "mean"
            ),
            Average_Frequency=(
                "frequency",
                "mean"
            ),
            Average_Recency=(
                "recency",
                "mean"
            )
        )
    )

else:

    segment_summary = pd.DataFrame()


# ------------------------------------------------------------
# Largest segment
# ------------------------------------------------------------

if not segment_summary.empty:

    largest_segment = (
        segment_summary
        .loc[
            segment_summary["Customers"].idxmax(),
            "Customer_Segment"
        ]
    )

else:

    largest_segment = "N/A"


# ------------------------------------------------------------
# Highest revenue segment
# ------------------------------------------------------------

if not segment_summary.empty:

    highest_revenue_segment = (
        segment_summary
        .loc[
            segment_summary["Revenue"].idxmax(),
            "Customer_Segment"
        ]
    )

else:

    highest_revenue_segment = "N/A"


# ============================================================
# STEP 11 — CHURN FINDINGS
# ============================================================

if {
    "Customer_Segment",
    "churn"
}.issubset(rfm.columns):

    segment_churn = (
        rfm
        .groupby(
            "Customer_Segment",
            as_index=False
        )
        .agg(
            Customers=(
                "customer_id",
                "nunique"
            ),
            Churned_Customers=(
                "churn",
                "sum"
            )
        )
    )

    segment_churn["Churn_Rate"] = np.where(
        segment_churn["Customers"] > 0,
        (
            segment_churn["Churned_Customers"] /
            segment_churn["Customers"]
        ) * 100,
        0
    )

    highest_churn_segment = (
        segment_churn
        .loc[
            segment_churn["Churn_Rate"].idxmax(),
            "Customer_Segment"
        ]
    )

    highest_segment_churn_rate = (
        segment_churn["Churn_Rate"].max()
    )

else:

    segment_churn = pd.DataFrame()

    highest_churn_segment = "N/A"
    highest_segment_churn_rate = 0


# ============================================================
# STEP 12 — HIGH-VALUE AT-RISK CUSTOMERS
# ============================================================

if {
    "R_Score",
    "M_Score",
    "churn"
}.issubset(rfm.columns):

    high_value_at_risk = rfm[
        (
            (rfm["R_Score"] <= 2)
            &
            (rfm["M_Score"] >= 4)
            &
            (rfm["churn"] == 0)
        )
    ].copy()

else:

    high_value_at_risk = pd.DataFrame()


high_value_at_risk_count = len(
    high_value_at_risk
)


# ------------------------------------------------------------
# Save high-value customers
# ------------------------------------------------------------

high_value_at_risk.to_csv(
    reports_folder /
    "High_Value_At_Risk_Customers.csv",
    index=False
)


# ============================================================
# STEP 13 — CREATE FINAL BUSINESS INSIGHTS TABLE
# ============================================================

final_business_insights = pd.DataFrame({

    "Category": [
        "Business",
        "Business",
        "Product",
        "Product",
        "Customer",
        "Customer",
        "Purchase Behavior",
        "Purchase Behavior",
        "Purchase Behavior",
        "Payment",
        "Churn",
        "Churn",
        "RFM"
    ],

    "Insight": [
        "Total Revenue",
        "Average Order Value",
        "Highest Revenue Category",
        "Highest Revenue Category Revenue",
        "Largest Customer Segment",
        "Highest Revenue Customer Segment",
        "Best Purchase Day",
        "Peak Purchase Hour",
        "Highest Revenue Month",
        "Most Used Payment Method",
        "Overall Churn Rate",
        "Highest Churn Segment",
        "High-Value At-Risk Customers"
    ],

    "Value": [
        total_revenue,
        average_order_value,
        best_category,
        best_category_revenue,
        largest_segment,
        highest_revenue_segment,
        best_purchase_day,
        f"{int(best_purchase_hour)}:00",
        best_month,
        most_used_payment,
        churn_rate,
        highest_churn_segment,
        high_value_at_risk_count
    ]
})


# ------------------------------------------------------------
# Save insights
# ------------------------------------------------------------

final_business_insights.to_csv(
    reports_folder /
    "Final_Business_Insights.csv",
    index=False
)


# ============================================================
# STEP 14 — SAVE SEGMENT SUMMARY
# ============================================================

if not segment_summary.empty:

    segment_summary.to_csv(
        reports_folder /
        "Final_RFM_Segment_Summary.csv",
        index=False
    )


# ============================================================
# STEP 15 — SAVE SEGMENT CHURN
# ============================================================

if not segment_churn.empty:

    segment_churn.to_csv(
        reports_folder /
        "Final_Segment_Churn.csv",
        index=False
    )


# ============================================================
# STEP 16 — SAVE PRODUCT PERFORMANCE
# ============================================================

category_performance.to_csv(
    reports_folder /
    "Final_Category_Performance.csv",
    index=False
)


# ============================================================
# STEP 17 — SAVE PAYMENT PERFORMANCE
# ============================================================

payment_performance.to_csv(
    reports_folder /
    "Final_Payment_Performance.csv",
    index=False
)


# ============================================================
# STEP 18 — SAVE MONTHLY PERFORMANCE
# ============================================================

monthly_sales.to_csv(
    reports_folder /
    "Final_Monthly_Sales.csv",
    index=False
)


# ============================================================
# STEP 19 — SAVE PURCHASE BEHAVIOR
# ============================================================

day_performance.to_csv(
    reports_folder /
    "Final_Day_Performance.csv",
    index=False
)

hour_performance.to_csv(
    reports_folder /
    "Final_Hour_Performance.csv",
    index=False
)

time_period_performance.to_csv(
    reports_folder /
    "Final_Time_Period_Performance.csv",
    index=False
)


# ============================================================
# STEP 20 — SAVE BUSINESS RECOMMENDATIONS
# ============================================================

# Make sure exactly 5 recommendations are included

recommendations = recommendations.head(5)

recommendations.to_csv(
    insights_folder /
    "Final_5_Business_Recommendations.csv",
    index=False
)


# ============================================================
# STEP 21 — CREATE FINAL PROJECT REPORT
# ============================================================

report_text = f"""
============================================================
CUSTOMER BEHAVIOR ANALYSIS
FINAL PROJECT REPORT
============================================================

Project:
Customer Behavior Analysis for E-Commerce

Objective:
Analyze customer purchasing behavior, identify valuable
customer segments, understand purchase patterns, analyze
churn and provide actionable business recommendations.


============================================================
1. DATASET OVERVIEW
============================================================

Total Customers:
{total_customers:,}

Total Transactions:
{total_transactions:,}

Total Revenue:
{total_revenue:,.2f}

Average Order Value:
{average_order_value:,.2f}

Total Quantity Sold:
{total_quantity:,}


============================================================
2. CUSTOMER BEHAVIOR
============================================================

Largest Customer Segment:
{largest_segment}

Highest Revenue Customer Segment:
{highest_revenue_segment}

Average Customer Spend:
{customer_kpis["Total_Spend"].mean()
 if "Total_Spend" in customer_kpis.columns else 0:,.2f}

Average Purchase Frequency:
{customer_kpis["Purchase_Frequency"].mean()
 if "Purchase_Frequency" in customer_kpis.columns else 0:,.2f}


============================================================
3. PRODUCT ANALYSIS
============================================================

Highest Revenue Product Category:
{best_category}

Revenue Generated by Category:
{best_category_revenue:,.2f}


============================================================
4. PURCHASE PATTERN ANALYSIS
============================================================

Best Purchase Day:
{best_purchase_day}

Peak Purchase Hour:
{int(best_purchase_hour)}:00

Highest Revenue Month:
{best_month}

Highest Revenue Month Value:
{best_month_revenue:,.2f}


============================================================
5. PAYMENT ANALYSIS
============================================================

Most Used Payment Method:
{most_used_payment}


============================================================
6. CHURN ANALYSIS
============================================================

Total Churned Customers:
{churned_customers:,}

Overall Churn Rate:
{churn_rate:.2f}%

Highest Churn Segment:
{highest_churn_segment}

Highest Segment Churn Rate:
{highest_segment_churn_rate:.2f}%

High-Value At-Risk Customers:
{high_value_at_risk_count:,}


============================================================
7. RFM ANALYSIS
============================================================

RFM analysis was used to classify customers based on:

R — Recency
F — Frequency
M — Monetary Value

The segmentation helps identify:

• High-value customers
• Loyal customers
• Potential loyal customers
• New customers
• At-risk customers
• Customers requiring attention
• Hibernating customers


============================================================
8. TOP 5 BUSINESS RECOMMENDATIONS
============================================================

1. RETAIN HIGH-VALUE AT-RISK CUSTOMERS

Target high-value customers showing declining recent
activity with personalized retention campaigns, offers,
reminders and loyalty benefits.


2. USE RFM-BASED PERSONALIZATION

Use RFM segments to create different marketing strategies
for Champions, Loyal Customers, New Customers, At-Risk
Customers and other customer groups.


3. FOCUS ON HIGH-REVENUE PRODUCTS

Prioritize the highest-revenue product category for
inventory planning, promotions, cross-selling and
product bundling.


4. OPTIMIZE PROMOTIONAL TIMING

Schedule campaigns around the strongest purchase days,
hours and periods to improve campaign engagement and
conversion.


5. IMPLEMENT EARLY CHURN DETECTION

Monitor recency and purchase frequency to identify
customers whose engagement is declining and target them
with proactive retention campaigns.


============================================================
9. EXPECTED BUSINESS BENEFITS
============================================================

• Improved customer retention
• Better customer segmentation
• More targeted marketing
• Improved product planning
• Better promotional timing
• Reduced customer churn
• Improved customer lifetime value
• Data-driven decision making


============================================================
10. TOOLS USED
============================================================

Python
Pandas
NumPy
Matplotlib
Seaborn
Jupyter Notebook / Google Colab
Power BI
SQL
Excel


============================================================
11. PROJECT WORKFLOW
============================================================

Phase 1:
Data Loading and Exploratory Data Analysis

Phase 2:
Data Cleaning and Feature Engineering

Phase 3:
RFM Analysis and Customer Segmentation

Phase 4:
Purchase Pattern Analysis

Phase 5:
Churn Analysis

Phase 6:
Data Visualization

Phase 7:
Power BI Dashboard Preparation

Phase 8:
Business Insights and Recommendations

Phase 9:
Final Report and Project Deliverables


============================================================
END OF FINAL PROJECT REPORT
============================================================
"""


# ============================================================
# STEP 22 — SAVE FINAL REPORT
# ============================================================

with open(
    reports_folder / "Final_Project_Report.txt",
    "w",
    encoding="utf-8"
) as file:

    file.write(report_text)


# ============================================================
# STEP 23 — CREATE README FILE
# ============================================================

readme_text = f"""
# Customer Behavior Analysis

## Project Overview

This project analyzes e-commerce customer behavior using
Python, Pandas, NumPy, Matplotlib, Seaborn and Power BI.

## Objective

The main objectives are:

- Understand customer purchasing behavior
- Analyze sales and product performance
- Perform RFM analysis
- Segment customers
- Identify churn patterns
- Identify high-value at-risk customers
- Generate actionable business recommendations

## Dataset

The project uses an e-commerce customer behavior dataset.

## Project Phases

### Step 1
Data Loading and Understanding

### Step 2
Data Cleaning and Feature Engineering

### Step 3
RFM Analysis and Customer Segmentation

### Step 4
Purchase Pattern Analysis

### Step 5
Churn Analysis

### Step 6
Visualization

### Step 7
Power BI Dashboard Preparation

### Step 8
Business Insights and Recommendations

### Step 9
Final Report and Deliverables

## Key KPIs

Total Customers: {total_customers:,}

Total Transactions: {total_transactions:,}

Total Revenue: {total_revenue:,.2f}

Average Order Value: {average_order_value:,.2f}

Churn Rate: {churn_rate:.2f}%

## Key Findings

Highest Revenue Category:
{best_category}

Largest Customer Segment:
{largest_segment}

Highest Revenue Segment:
{highest_revenue_segment}

Best Purchase Day:
{best_purchase_day}

Peak Purchase Hour:
{int(best_purchase_hour)}:00

Most Used Payment Method:
{most_used_payment}

High-Value At-Risk Customers:
{high_value_at_risk_count:,}

## Technologies

- Python
- Pandas
- NumPy
- Matplotlib
- Seaborn
- Power BI
- Jupyter Notebook
- Google Colab

## Dashboard Pages

1. Executive Overview
2. Customer Behavior
3. RFM Segmentation
4. Churn Analysis
5. Product & Purchase Patterns

## Business Recommendations

1. Retain high-value at-risk customers.
2. Use RFM-based personalization.
3. Focus on high-revenue products.
4. Optimize promotional timing.
5. Implement early churn detection.

## Author
Ankit Kumar Mahto
B.Tech CSE Student
Data Analytics Project
"""


# ============================================================
# STEP 24 — SAVE README
# ============================================================

with open(
    project_folder / "README.md",
    "w",
    encoding="utf-8"
) as file:

    file.write(readme_text)


# ============================================================
# STEP 25 — CREATE PROJECT METADATA
# ============================================================

metadata = pd.DataFrame({
    "Project_Property": [
        "Project Name",
        "Project Type",
        "Domain",
        "Primary Tools",
        "Dataset Type",
        "Total Customers",
        "Total Transactions",
        "Total Revenue",
        "Churn Rate",
        "Completion Date"
    ],
    "Value": [
        "Customer Behavior Analysis",
        "Data Analytics",
        "E-Commerce",
        "Python + Power BI",
        "E-Commerce Customer Behavior",
        total_customers,
        total_transactions,
        total_revenue,
        churn_rate,
        datetime.now().strftime("%Y-%m-%d")
    ]
})

metadata.to_csv(
    project_folder /
    "Project_Metadata.csv",
    index=False
)


# ============================================================
# STEP 26 — CREATE FINAL FILE INVENTORY
# ============================================================

file_inventory = []

for file in project_folder.rglob("*"):

    if file.is_file():

        file_inventory.append({
            "File_Name": file.name,
            "Folder": str(
                file.parent.relative_to(
                    project_folder
                )
            ),
            "File_Size_KB": round(
                file.stat().st_size / 1024,
                2
            )
        })


file_inventory = pd.DataFrame(
    file_inventory
)

file_inventory.to_csv(
    project_folder /
    "Project_File_Inventory.csv",
    index=False
)


# ============================================================
# STEP 27 — CREATE FINAL ZIP FILE
# ============================================================

zip_path = shutil.make_archive(
    "Customer_Behavior_Analysis_Final",
    "zip",
    root_dir=project_folder
)


# ============================================================
# STEP 28 — DISPLAY FINAL KPIs
# ============================================================

print("\n====================================================")
print("FINAL PROJECT KPIs")
print("====================================================\n")

display(final_kpis)


# ============================================================
# STEP 29 — DISPLAY FINAL BUSINESS INSIGHTS
# ============================================================

print("\n====================================================")
print("FINAL BUSINESS INSIGHTS")
print("====================================================\n")

display(final_business_insights)


# ============================================================
# STEP 30 — DISPLAY EXACTLY 5 RECOMMENDATIONS
# ============================================================

print("\n====================================================")
print("FINAL 5 BUSINESS RECOMMENDATIONS")
print("====================================================\n")

display(recommendations)


# ============================================================
# STEP 31 — DISPLAY PROJECT FILE INVENTORY
# ============================================================

print("\n====================================================")
print("PROJECT FILE INVENTORY")
print("====================================================\n")

display(file_inventory)


# ============================================================
# STEP 32 — FINAL PROJECT COMPLETION MESSAGE
# ============================================================

print("\n====================================================")
print("STEP 9 COMPLETED SUCCESSFULLY")
print("====================================================")

print("\nFinal Project Folder:")
print(project_folder)

print("\nFinal ZIP File:")
print(zip_path)

print("\nProject deliverables include:")

print("✓ Cleaned transaction dataset")
print("✓ Customer-level feature dataset")
print("✓ RFM segmentation dataset")
print("✓ Power BI-ready datasets")
print("✓ KPI report")
print("✓ Business insights")
print("✓ Churn analysis")
print("✓ RFM analysis")
print("✓ 5 business recommendations")
print("✓ Final project report")
print("✓ README file")
print("✓ Complete ZIP package")

print("\n====================================================")
print("CUSTOMER BEHAVIOR ANALYSIS PROJECT COMPLETED")
print("====================================================")
```
