```python
# ============================================================
#               PURCHASE PATTERN ANALYSIS
# ============================================================

# ------------------------------------------------------------
# 1. IMPORT REQUIRED LIBRARIES
# ------------------------------------------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


# ------------------------------------------------------------
# 2. LOAD CLEANED TRANSACTION DATA
# ------------------------------------------------------------

# Load the cleaned transaction-level dataset created in Phase 2
data = pd.read_csv("ecommerce_cleaned_transactions.csv")

# Convert purchase date back to datetime format
data["purchase_date"] = pd.to_datetime(
    data["purchase_date"],
    errors="coerce"
)

# Display basic information
print("Dataset Shape:", data.shape)

print("\nFirst 5 Records:")
display(data.head())


# ------------------------------------------------------------
# 3. BASIC PURCHASE STATISTICS
# ------------------------------------------------------------

print("\nPurchase Statistics:")
display(
    data[
        [
            "product_price",
            "quantity",
            "total_purchase_amount"
        ]
    ].describe()
)


# ------------------------------------------------------------
# 4. TOTAL SALES / REVENUE
# ------------------------------------------------------------

total_revenue = data["total_purchase_amount"].sum()

print(
    "Total Revenue:",
    round(total_revenue, 2)
)


# ------------------------------------------------------------
# 5. TOTAL NUMBER OF TRANSACTIONS
# ------------------------------------------------------------

total_transactions = data["customer_id"].count()

print(
    "Total Transactions:",
    total_transactions
)


# ------------------------------------------------------------
# 6. TOTAL UNIQUE CUSTOMERS
# ------------------------------------------------------------

unique_customers = data["customer_id"].nunique()

print(
    "Unique Customers:",
    unique_customers
)


# ------------------------------------------------------------
# 7. TOTAL QUANTITY SOLD
# ------------------------------------------------------------

total_quantity = data["quantity"].sum()

print(
    "Total Quantity Sold:",
    total_quantity
)


# ------------------------------------------------------------
# 8. AVERAGE ORDER VALUE
# ------------------------------------------------------------

average_order_value = data[
    "total_purchase_amount"
].mean()

print(
    "Average Transaction Value:",
    round(average_order_value, 2)
)


# ============================================================
# 9. PRODUCT CATEGORY ANALYSIS
# ============================================================

# ------------------------------------------------------------
# 9A. CATEGORY-WISE TRANSACTIONS
# ------------------------------------------------------------

category_transactions = (
    data["product_category"]
    .value_counts()
    .reset_index()
)

category_transactions.columns = [
    "Product_Category",
    "Transaction_Count"
]

print("\nTransactions by Product Category:")
display(category_transactions)


# ------------------------------------------------------------
# 9B. CATEGORY-WISE REVENUE
# ------------------------------------------------------------

category_revenue = (
    data
    .groupby("product_category")
    .agg(
        Transactions=("customer_id", "count"),
        Quantity_Sold=("quantity", "sum"),
        Total_Revenue=("total_purchase_amount", "sum"),
        Average_Order_Value=("total_purchase_amount", "mean")
    )
    .reset_index()
)

category_revenue = category_revenue.sort_values(
    "Total_Revenue",
    ascending=False
)

print("\nProduct Category Performance:")
display(category_revenue)


# ------------------------------------------------------------
# 9C. CATEGORY REVENUE PERCENTAGE
# ------------------------------------------------------------

category_revenue["Revenue_Percentage"] = (
    category_revenue["Total_Revenue"]
    / category_revenue["Total_Revenue"].sum()
    * 100
)

category_revenue = category_revenue.round(2)

print("\nCategory Revenue Contribution:")
display(category_revenue)


# ------------------------------------------------------------
# 9D. VISUALIZE CATEGORY REVENUE
# ------------------------------------------------------------

plt.figure(figsize=(6,3))

sns.barplot(
    data=category_revenue,
    x="Total_Revenue",
    y="product_category"
)

plt.title("Revenue by Product Category")
plt.xlabel("Total Revenue")
plt.ylabel("Product Category")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 9E. VISUALIZE CATEGORY QUANTITY
# ------------------------------------------------------------

plt.figure(figsize=(6,3))

sns.barplot(
    data=category_revenue.sort_values(
        "Quantity_Sold",
        ascending=False
    ),
    x="Quantity_Sold",
    y="product_category"
)

plt.title("Quantity Sold by Product Category")
plt.xlabel("Quantity Sold")
plt.ylabel("Product Category")
plt.tight_layout()
plt.show()


# ============================================================
# 10. PAYMENT METHOD ANALYSIS
# ============================================================

# ------------------------------------------------------------
# 10A. PAYMENT METHOD USAGE
# ------------------------------------------------------------

payment_usage = (
    data["payment_method"]
    .value_counts()
    .reset_index()
)

payment_usage.columns = [
    "Payment_Method",
    "Transaction_Count"
]

print("\nPayment Method Usage:")
display(payment_usage)


# ------------------------------------------------------------
# 10B. PAYMENT METHOD REVENUE
# ------------------------------------------------------------

payment_analysis = (
    data
    .groupby("payment_method")
    .agg(
        Transactions=("customer_id", "count"),
        Total_Revenue=("total_purchase_amount", "sum"),
        Average_Order_Value=("total_purchase_amount", "mean")
    )
    .reset_index()
)

payment_analysis["Revenue_Percentage"] = (
    payment_analysis["Total_Revenue"]
    / payment_analysis["Total_Revenue"].sum()
    * 100
)

payment_analysis = payment_analysis.sort_values(
    "Total_Revenue",
    ascending=False
)

payment_analysis = payment_analysis.round(2)

print("\nPayment Method Analysis:")
display(payment_analysis)


# ------------------------------------------------------------
# 10C. PAYMENT METHOD REVENUE VISUALIZATION
# ------------------------------------------------------------

plt.figure(figsize=(6,3))

sns.barplot(
    data=payment_analysis,
    x="Total_Revenue",
    y="payment_method"
)

plt.title("Revenue by Payment Method")
plt.xlabel("Total Revenue")
plt.ylabel("Payment Method")
plt.tight_layout()
plt.show()


# ============================================================
# 11. MONTHLY PURCHASE ANALYSIS
# ============================================================

# ------------------------------------------------------------
# 11A. CREATE YEAR AND MONTH
# ------------------------------------------------------------

data["purchase_year"] = (
    data["purchase_date"].dt.year
)

data["purchase_month"] = (
    data["purchase_date"].dt.month
)

data["year_month"] = (
    data["purchase_date"]
    .dt.to_period("M")
    .astype(str)
)


# ------------------------------------------------------------
# 11B. MONTHLY SALES
# ------------------------------------------------------------

monthly_sales = (
    data
    .groupby("year_month")
    .agg(
        Transactions=("customer_id", "count"),
        Unique_Customers=("customer_id", "nunique"),
        Quantity_Sold=("quantity", "sum"),
        Revenue=("total_purchase_amount", "sum")
    )
    .reset_index()
)

print("\nMonthly Purchase Analysis:")
display(monthly_sales.head(20))


# ------------------------------------------------------------
# 11C. MONTHLY REVENUE TREND
# ------------------------------------------------------------

plt.figure(figsize=(10,5))

sns.lineplot(
    data=monthly_sales,
    x="year_month",
    y="Revenue",
    marker="o"
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 11D. MONTHLY TRANSACTION TREND
# ------------------------------------------------------------

plt.figure(figsize=(8,4))

sns.lineplot(
    data=monthly_sales,
    x="year_month",
    y="Transactions",
    marker="o"
)

plt.title("Monthly Transaction Trend")
plt.xlabel("Month")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()


# ============================================================
# 12. YEARLY PURCHASE ANALYSIS
# ============================================================

yearly_sales = (
    data
    .groupby("purchase_year")
    .agg(
        Transactions=("customer_id", "count"),
        Unique_Customers=("customer_id", "nunique"),
        Quantity_Sold=("quantity", "sum"),
        Revenue=("total_purchase_amount", "sum")
    )
    .reset_index()
)

print("\nYearly Purchase Analysis:")
display(yearly_sales)


# ------------------------------------------------------------
# 12A. YEARLY REVENUE VISUALIZATION
# ------------------------------------------------------------

plt.figure(figsize=(6,3))

sns.barplot(
    data=yearly_sales,
    x="purchase_year",
    y="Revenue"
)

plt.title("Yearly Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.tight_layout()
plt.show()


# ============================================================
# 13. DAY-OF-WEEK PURCHASE PATTERN
# ============================================================

# ------------------------------------------------------------
# 13A. CREATE DAY OF WEEK
# ------------------------------------------------------------

data["day_of_week"] = (
    data["purchase_date"].dt.day_name()
)

# Correct chronological order
day_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]


# ------------------------------------------------------------
# 13B. DAY-WISE ANALYSIS
# ------------------------------------------------------------

day_analysis = (
    data
    .groupby("day_of_week")
    .agg(
        Transactions=("customer_id", "count"),
        Revenue=("total_purchase_amount", "sum"),
        Quantity_Sold=("quantity", "sum")
    )
    .reindex(day_order)
    .reset_index()
)

print("\nDay-of-Week Purchase Pattern:")
display(day_analysis)


# ------------------------------------------------------------
# 13C. DAY-WISE TRANSACTIONS
# ------------------------------------------------------------

plt.figure(figsize=(6,3))

sns.barplot(
    data=day_analysis,
    x="day_of_week",
    y="Transactions"
)

plt.title("Transactions by Day of Week")
plt.xlabel("Day")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 13D. DAY-WISE REVENUE
# ------------------------------------------------------------

plt.figure(figsize=(6,3))

sns.barplot(
    data=day_analysis,
    x="day_of_week",
    y="Revenue"
)

plt.title("Revenue by Day of Week")
plt.xlabel("Day")
plt.ylabel("Revenue")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()


# ============================================================
# 14. HOURLY PURCHASE PATTERN
# ============================================================

# ------------------------------------------------------------
# 14A. EXTRACT PURCHASE HOUR
# ------------------------------------------------------------

data["purchase_hour"] = (
    data["purchase_date"].dt.hour
)


# ------------------------------------------------------------
# 14B. HOURLY ANALYSIS
# ------------------------------------------------------------

hourly_analysis = (
    data
    .groupby("purchase_hour")
    .agg(
        Transactions=("customer_id", "count"),
        Revenue=("total_purchase_amount", "sum"),
        Quantity_Sold=("quantity", "sum")
    )
    .reset_index()
)

print("\nHourly Purchase Pattern:")
display(hourly_analysis)


# ------------------------------------------------------------
# 14C. HOURLY TRANSACTION TREND
# ------------------------------------------------------------

plt.figure(figsize=(8,4))

sns.lineplot(
    data=hourly_analysis,
    x="purchase_hour",
    y="Transactions",
    marker="o"
)

plt.title("Transactions by Purchase Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Number of Transactions")
plt.xticks(range(24))
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 14D. HOURLY REVENUE TREND
# ------------------------------------------------------------

plt.figure(figsize=(8,5))

sns.lineplot(
    data=hourly_analysis,
    x="purchase_hour",
    y="Revenue",
    marker="o"
)

plt.title("Revenue by Purchase Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Revenue")
plt.xticks(range(24))
plt.tight_layout()
plt.show()


# ============================================================
# 15. AGE GROUP ANALYSIS
# ============================================================

# ------------------------------------------------------------
# 15A. CREATE AGE GROUPS
# ------------------------------------------------------------

data["age_group"] = pd.cut(
    data["customer_age"],
    bins=[0, 18, 25, 35, 45, 55, 65, 100],
    labels=[
        "Under 18",
        "18-25",
        "26-35",
        "36-45",
        "46-55",
        "56-65",
        "65+"
    ],
    right=True
)


# ------------------------------------------------------------
# 15B. AGE GROUP PERFORMANCE
# ------------------------------------------------------------

age_analysis = (
    data
    .groupby("age_group", observed=True)
    .agg(
        Transactions=("customer_id", "count"),
        Unique_Customers=("customer_id", "nunique"),
        Quantity_Sold=("quantity", "sum"),
        Revenue=("total_purchase_amount", "sum"),
        Average_Order_Value=("total_purchase_amount", "mean")
    )
    .reset_index()
)

print("\nAge Group Purchase Analysis:")
display(age_analysis)


# ------------------------------------------------------------
# 15C. AGE GROUP REVENUE
# ------------------------------------------------------------

plt.figure(figsize=(6,3))

sns.barplot(
    data=age_analysis,
    x="age_group",
    y="Revenue"
)

plt.title("Revenue by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Revenue")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()


# ============================================================
# 16. GENDER-WISE PURCHASE ANALYSIS
# ============================================================

gender_analysis = (
    data
    .groupby("gender")
    .agg(
        Transactions=("customer_id", "count"),
        Unique_Customers=("customer_id", "nunique"),
        Quantity_Sold=("quantity", "sum"),
        Revenue=("total_purchase_amount", "sum"),
        Average_Order_Value=("total_purchase_amount", "mean")
    )
    .reset_index()
)

print("\nGender-wise Purchase Analysis:")
display(gender_analysis)


# ------------------------------------------------------------
# 16A. GENDER REVENUE VISUALIZATION
# ------------------------------------------------------------

plt.figure(figsize=(6,3))

sns.barplot(
    data=gender_analysis,
    x="gender",
    y="Revenue"
)

plt.title("Revenue by Gender")
plt.xlabel("Gender")
plt.ylabel("Revenue")
plt.tight_layout()
plt.show()


# ============================================================
# 17. PRODUCT CATEGORY × GENDER ANALYSIS
# ============================================================

category_gender = (
    data
    .pivot_table(
        index="product_category",
        columns="gender",
        values="total_purchase_amount",
        aggfunc="sum"
    )
)

print("\nRevenue by Product Category and Gender:")
display(category_gender)


# ------------------------------------------------------------
# 17A. HEATMAP
# ------------------------------------------------------------

plt.figure(figsize=(6,3))

sns.heatmap(
    category_gender,
    annot=True,
    fmt=".0f"
)

plt.title("Revenue: Product Category vs Gender")
plt.xlabel("Gender")
plt.ylabel("Product Category")
plt.tight_layout()
plt.show()


# ============================================================
# 18. PRODUCT CATEGORY × AGE GROUP
# ============================================================

category_age = (
    data
    .pivot_table(
        index="product_category",
        columns="age_group",
        values="total_purchase_amount",
        aggfunc="sum",
        observed=False
    )
)

print("\nRevenue by Product Category and Age Group:")
display(category_age)


# ------------------------------------------------------------
# 18A. HEATMAP
# ------------------------------------------------------------

plt.figure(figsize=(8,4))

sns.heatmap(
    category_age,
    annot=True,
    fmt=".0f"
)

plt.title("Revenue: Product Category vs Age Group")
plt.xlabel("Age Group")
plt.ylabel("Product Category")
plt.tight_layout()
plt.show()


# ============================================================
# 19. PRODUCT CATEGORY × PAYMENT METHOD
# ============================================================

category_payment = (
    data
    .pivot_table(
        index="product_category",
        columns="payment_method",
        values="total_purchase_amount",
        aggfunc="sum"
    )
)

print("\nRevenue by Product Category and Payment Method:")
display(category_payment)


# ------------------------------------------------------------
# 19A. HEATMAP
# ------------------------------------------------------------

plt.figure(figsize=(8,4))

sns.heatmap(
    category_payment,
    annot=True,
    fmt=".0f"
)

plt.title("Revenue: Product Category vs Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Product Category")
plt.tight_layout()
plt.show()


# ============================================================
# 20. PURCHASE QUANTITY ANALYSIS
# ============================================================

quantity_analysis = (
    data
    .groupby("quantity")
    .agg(
        Transactions=("customer_id", "count"),
        Revenue=("total_purchase_amount", "sum")
    )
    .reset_index()
)

print("\nQuantity-wise Purchase Analysis:")
display(quantity_analysis)


# ------------------------------------------------------------
# 20A. QUANTITY DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(6,3))

sns.barplot(
    data=quantity_analysis,
    x="quantity",
    y="Transactions"
)

plt.title("Transactions by Quantity Purchased")
plt.xlabel("Quantity")
plt.ylabel("Transactions")
plt.tight_layout()
plt.show()


# ============================================================
# 21. PRICE RANGE ANALYSIS
# ============================================================

# ------------------------------------------------------------
# 21A. CREATE PRODUCT PRICE BANDS
# ------------------------------------------------------------

data["price_range"] = pd.cut(
    data["product_price"],
    bins=[0, 100, 500, 1000, 2500, 5000, np.inf],
    labels=[
        "₹0-100",
        "₹101-500",
        "₹501-1000",
        "₹1001-2500",
        "₹2501-5000",
        "₹5000+"
    ]
)


# ------------------------------------------------------------
# 21B. PRICE RANGE PERFORMANCE
# ------------------------------------------------------------

price_analysis = (
    data
    .groupby("price_range", observed=True)
    .agg(
        Transactions=("customer_id", "count"),
        Quantity_Sold=("quantity", "sum"),
        Revenue=("total_purchase_amount", "sum")
    )
    .reset_index()
)

print("\nPrice Range Analysis:")
display(price_analysis)


# ------------------------------------------------------------
# 21C. PRICE RANGE REVENUE
# ------------------------------------------------------------

plt.figure(figsize=(6,3))

sns.barplot(
    data=price_analysis,
    x="price_range",
    y="Revenue"
)

plt.title("Revenue by Product Price Range")
plt.xlabel("Price Range")
plt.ylabel("Revenue")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()


# ============================================================
# 22. RETURN / REFUND PATTERN
# ============================================================

# ------------------------------------------------------------
# 22A. RETURN STATUS
# ------------------------------------------------------------

# Convert Returns into a binary return indicator
data["return_flag"] = (
    data["returns"]
    .fillna(0)
    .astype(int)
)

return_analysis = (
    data
    .groupby("return_flag")
    .agg(
        Transactions=("customer_id", "count"),
        Revenue=("total_purchase_amount", "sum")
    )
    .reset_index()
)

print("\nReturn Analysis:")
display(return_analysis)


# ------------------------------------------------------------
# 22B. OVERALL RETURN RATE
# ------------------------------------------------------------

return_rate = (
    data["return_flag"].mean() * 100
)

print(
    "Overall Return Rate:",
    round(return_rate, 2),
    "%"
)


# ------------------------------------------------------------
# 22C. RETURN RATE BY CATEGORY
# ------------------------------------------------------------

category_returns = (
    data
    .groupby("product_category")
    .agg(
        Transactions=("customer_id", "count"),
        Returns=("return_flag", "sum"),
        Return_Rate=("return_flag", "mean")
    )
    .reset_index()
)

category_returns["Return_Rate"] = (
    category_returns["Return_Rate"] * 100
)

category_returns = category_returns.round(2)

print("\nReturn Rate by Product Category:")
display(category_returns)


# ------------------------------------------------------------
# 22D. RETURN RATE VISUALIZATION
# ------------------------------------------------------------

plt.figure(figsize=(6,3))

sns.barplot(
    data=category_returns,
    x="Return_Rate",
    y="product_category"
)

plt.title("Return Rate by Product Category")
plt.xlabel("Return Rate (%)")
plt.ylabel("Product Category")
plt.tight_layout()
plt.show()


# ============================================================
# 23. TOP CUSTOMERS BY TOTAL SPENDING
# ============================================================

top_customers = (
    data
    .groupby(
        ["customer_id", "customer_name"],
        as_index=False
    )
    .agg(
        Total_Spending=("total_purchase_amount", "sum"),
        Purchase_Frequency=("customer_id", "count"),
        Total_Quantity=("quantity", "sum")
    )
    .sort_values(
        "Total_Spending",
        ascending=False
    )
)

print("\nTop 20 Customers by Spending:")
display(
    top_customers.head(20)
)


# ============================================================
# 24. TOP CUSTOMERS BY PURCHASE FREQUENCY
# ============================================================

frequent_customers = (
    top_customers
    .sort_values(
        "Purchase_Frequency",
        ascending=False
    )
)

print("\nTop 20 Customers by Purchase Frequency:")
display(
    frequent_customers.head(20)
)


# ============================================================
# 25. CATEGORY-CUSTOMER PURCHASE MATRIX
# ============================================================

customer_category = (
    data
    .pivot_table(
        index="customer_id",
        columns="product_category",
        values="total_purchase_amount",
        aggfunc="sum",
        fill_value=0
    )
)

print("\nCustomer × Product Category Matrix:")
display(
    customer_category.head()
)


# ============================================================
# 26. FIND MOST POPULAR CATEGORY
# ============================================================

most_popular_category = (
    data["product_category"]
    .value_counts()
    .idxmax()
)

print(
    "Most Popular Product Category:",
    most_popular_category
)


# ============================================================
# 27. FIND HIGHEST-REVENUE CATEGORY
# ============================================================

highest_revenue_category = (
    category_revenue
    .sort_values(
        "Total_Revenue",
        ascending=False
    )
    .iloc[0]["product_category"]
)

print(
    "Highest Revenue Category:",
    highest_revenue_category
)


# ============================================================
# 28. FIND MOST USED PAYMENT METHOD
# ============================================================

most_used_payment = (
    data["payment_method"]
    .value_counts()
    .idxmax()
)

print(
    "Most Used Payment Method:",
    most_used_payment
)


# ============================================================
# 29. FIND PEAK PURCHASE DAY
# ============================================================

peak_purchase_day = (
    day_analysis
    .sort_values(
        "Transactions",
        ascending=False
    )
    .iloc[0]["day_of_week"]
)

print(
    "Peak Purchase Day:",
    peak_purchase_day
)


# ============================================================
# 30. FIND PEAK PURCHASE HOUR
# ============================================================

peak_purchase_hour = (
    hourly_analysis
    .sort_values(
        "Transactions",
        ascending=False
    )
    .iloc[0]["purchase_hour"]
)

print(
    "Peak Purchase Hour:",
    peak_purchase_hour
)


# ============================================================
# 31. FIND BEST REVENUE MONTH
# ============================================================

best_revenue_month = (
    monthly_sales
    .sort_values(
        "Revenue",
        ascending=False
    )
    .iloc[0]
)

print(
    "Highest Revenue Month:",
    best_revenue_month["year_month"]
)

print(
    "Revenue:",
    round(best_revenue_month["Revenue"], 2)
)


# ============================================================
# 32. SAVE PHASE 4 ANALYSIS FILES
# ============================================================

# Save category analysis
category_revenue.to_csv(
    "category_purchase_analysis.csv",
    index=False
)

# Save payment analysis
payment_analysis.to_csv(
    "payment_method_analysis.csv",
    index=False
)

# Save monthly analysis
monthly_sales.to_csv(
    "monthly_purchase_analysis.csv",
    index=False
)

# Save yearly analysis
yearly_sales.to_csv(
    "yearly_purchase_analysis.csv",
    index=False
)

# Save day analysis
day_analysis.to_csv(
    "day_of_week_analysis.csv",
    index=False
)

# Save hourly analysis
hourly_analysis.to_csv(
    "hourly_purchase_analysis.csv",
    index=False
)

# Save age analysis
age_analysis.to_csv(
    "age_group_purchase_analysis.csv",
    index=False
)

# Save gender analysis
gender_analysis.to_csv(
    "gender_purchase_analysis.csv",
    index=False
)

# Save return analysis
category_returns.to_csv(
    "category_return_analysis.csv",
    index=False
)

# Save top customers
top_customers.to_csv(
    "top_customers_analysis.csv",
    index=False
)


# ============================================================
# 33. SAVE ENRICHED TRANSACTION DATA
# ============================================================

# Save the transaction dataset with the newly created
# purchase-pattern features
data.to_csv(
    "purchase_pattern_analysis_data.csv",
    index=False
)


# ============================================================
# 34. FINAL STEP 4 SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("STEP 4: PURCHASE PATTERN ANALYSIS COMPLETED")
print("=" * 60)

print("\nKey Analysis Completed:")
print("✓ Overall purchase statistics")
print("✓ Product category analysis")
print("✓ Payment method analysis")
print("✓ Monthly purchase trends")
print("✓ Yearly purchase trends")
print("✓ Day-of-week purchase patterns")
print("✓ Hourly purchase patterns")
print("✓ Age group analysis")
print("✓ Gender-wise analysis")
print("✓ Category × Gender analysis")
print("✓ Category × Age analysis")
print("✓ Category × Payment analysis")
print("✓ Quantity analysis")
print("✓ Product price analysis")
print("✓ Return analysis")
print("✓ Top customer analysis")

print("\nImportant Findings:")
print(
    "Most Popular Category:",
    most_popular_category
)

print(
    "Highest Revenue Category:",
    highest_revenue_category
)

print(
    "Most Used Payment Method:",
    most_used_payment
)

print(
    "Peak Purchase Day:",
    peak_purchase_day
)

print(
    "Peak Purchase Hour:",
    peak_purchase_hour
)

print(
    "Overall Return Rate:",
    round(return_rate, 2),
    "%"
)

print("\nOutput files created:")
print("1. category_purchase_analysis.csv")
print("2. payment_method_analysis.csv")
print("3. monthly_purchase_analysis.csv")
print("4. yearly_purchase_analysis.csv")
print("5. day_of_week_analysis.csv")
print("6. hourly_purchase_analysis.csv")
print("7. age_group_purchase_analysis.csv")
print("8. gender_purchase_analysis.csv")
print("9. category_return_analysis.csv")
print("10. top_customers_analysis.csv")
print("11. purchase_pattern_analysis_data.csv")
```
