```python
# ============================================================
#            STEP 6 — VISUALIZATION & DASHBOARD
# ============================================================

# ------------------------------------------------------------
# STEP 1 — IMPORT LIBRARIES
# ------------------------------------------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

# ------------------------------------------------------------
# STEP 2 — LOAD REQUIRED DATASETS
# ------------------------------------------------------------

transactions = pd.read_csv("ecommerce_cleaned_transactions.csv")
customers = pd.read_csv("customer_level_features.csv")
rfm = pd.read_csv("customer_rfm_segments.csv")

# ------------------------------------------------------------
# STEP 3 — CONVERT DATE COLUMN
# ------------------------------------------------------------

transactions["purchase_date"] = pd.to_datetime(
    transactions["purchase_date"],
    errors="coerce"
)

# ------------------------------------------------------------
# STEP 4 — BASIC DATA CHECK
# ------------------------------------------------------------

print("Transactions Shape:", transactions.shape)
print("Customer-Level Shape:", customers.shape)
print("RFM Shape:", rfm.shape)

print("\nTransaction Columns:")
print(transactions.columns.tolist())

print("\nCustomer Columns:")
print(customers.columns.tolist())

print("\nRFM Columns:")
print(rfm.columns.tolist())


# ============================================================
# SECTION A — OVERALL BUSINESS KPIs
# ============================================================

# ------------------------------------------------------------
# STEP 5 — CALCULATE KEY BUSINESS KPIs
# ------------------------------------------------------------

total_transactions = len(transactions)

total_customers = transactions["customer_id"].nunique()

total_revenue = transactions["total_purchase_amount"].sum()

average_order_value = transactions["total_purchase_amount"].mean()

total_quantity = transactions["quantity"].sum()

average_quantity = transactions["quantity"].mean()

if "returns" in transactions.columns:
    total_returns = transactions["returns"].sum()
else:
    total_returns = 0

if "churn" in transactions.columns:
    churn_rate = transactions["churn"].mean() * 100
else:
    churn_rate = np.nan

# ------------------------------------------------------------
# STEP 6 — DISPLAY KPIs
# ------------------------------------------------------------

print("\n================ BUSINESS KPIs ================\n")

print("Total Transactions       :", total_transactions)
print("Total Customers          :", total_customers)
print("Total Revenue            :", round(total_revenue, 2))
print("Average Order Value      :", round(average_order_value, 2))
print("Total Quantity Sold      :", total_quantity)
print("Average Quantity/Order   :", round(average_quantity, 2))
print("Total Returns            :", total_returns)
print("Overall Churn Rate (%)   :", round(churn_rate, 2))


# ============================================================
# SECTION B — SALES TREND VISUALIZATION
# ============================================================

# ------------------------------------------------------------
# STEP 7 — MONTHLY SALES
# ------------------------------------------------------------

monthly_sales = (
    transactions
    .set_index("purchase_date")
    .resample("M")["total_purchase_amount"]
    .sum()
    .reset_index()
)

monthly_sales["Month"] = monthly_sales["purchase_date"].dt.strftime("%Y-%m")

# ------------------------------------------------------------
# STEP 8 — MONTHLY REVENUE LINE CHART
# ------------------------------------------------------------

plt.figure(figsize=(6,3))

plt.plot(
    monthly_sales["purchase_date"],
    monthly_sales["total_purchase_amount"],
    marker="o"
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# STEP 9 — YEARLY SALES
# ------------------------------------------------------------

yearly_sales = (
    transactions
    .assign(Year=transactions["purchase_date"].dt.year)
    .groupby("Year", as_index=False)["total_purchase_amount"]
    .sum()
)

plt.figure(figsize=(6,3))

sns.barplot(
    data=yearly_sales,
    x="Year",
    y="total_purchase_amount"
)

plt.title("Yearly Revenue")
plt.xlabel("Year")
plt.ylabel("Total Revenue")
plt.tight_layout()
plt.show()


# ============================================================
# SECTION C — PRODUCT CATEGORY VISUALIZATION
# ============================================================

# ------------------------------------------------------------
# STEP 10 — CATEGORY PERFORMANCE
# ------------------------------------------------------------

category_summary = (
    transactions
    .groupby("product_category")
    .agg(
        Revenue=("total_purchase_amount", "sum"),
        Transactions=("customer_id", "count"),
        Quantity=("quantity", "sum"),
        Average_Order_Value=("total_purchase_amount", "mean")
    )
    .reset_index()
    .sort_values("Revenue", ascending=False)
)

print("\nCategory Performance:")
display(category_summary)


# ------------------------------------------------------------
# STEP 11 — CATEGORY REVENUE CHART
# ------------------------------------------------------------

plt.figure(figsize=(6,3))

sns.barplot(
    data=category_summary,
    x="product_category",
    y="Revenue"
)

plt.title("Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Revenue")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# STEP 12 — CATEGORY TRANSACTION DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(6,3))

sns.barplot(
    data=category_summary,
    x="product_category",
    y="Transactions"
)

plt.title("Transactions by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()


# ============================================================
# SECTION D — PAYMENT METHOD ANALYSIS
# ============================================================

# ------------------------------------------------------------
# STEP 13 — PAYMENT METHOD SUMMARY
# ------------------------------------------------------------

payment_summary = (
    transactions
    .groupby("payment_method")
    .agg(
        Transactions=("customer_id", "count"),
        Revenue=("total_purchase_amount", "sum"),
        Average_Order_Value=("total_purchase_amount", "mean")
    )
    .reset_index()
    .sort_values("Revenue", ascending=False)
)

print("\nPayment Method Performance:")
display(payment_summary)


# ------------------------------------------------------------
# STEP 14 — PAYMENT METHOD CHART
# ------------------------------------------------------------

plt.figure(figsize=(6,3))

sns.barplot(
    data=payment_summary,
    x="payment_method",
    y="Transactions"
)

plt.title("Transactions by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()


# ============================================================
# SECTION E — CUSTOMER SEGMENTATION VISUALIZATION
# ============================================================

# ------------------------------------------------------------
# STEP 15 — CUSTOMER SEGMENT SUMMARY
# ------------------------------------------------------------

if "Customer_Segment" in rfm.columns:

    segment_summary = (
        rfm
        .groupby("Customer_Segment")
        .agg(
            Customers=("customer_id", "count"),
            Revenue=("monetary", "sum"),
            Average_Spend=("monetary", "mean"),
            Average_Frequency=("frequency", "mean"),
            Average_Recency=("recency", "mean")
        )
        .reset_index()
        .sort_values("Revenue", ascending=False)
    )

    print("\nCustomer Segment Summary:")
    display(segment_summary)

else:
    segment_summary = pd.DataFrame()


# ------------------------------------------------------------
# STEP 16 — CUSTOMER SEGMENT DISTRIBUTION
# ------------------------------------------------------------

if not segment_summary.empty:

    plt.figure(figsize=(6,3))

    sns.barplot(
        data=segment_summary.sort_values(
            "Customers",
            ascending=False
        ),
        x="Customer_Segment",
        y="Customers"
    )

    plt.title("Customer Distribution by Segment")
    plt.xlabel("Customer Segment")
    plt.ylabel("Number of Customers")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()


# ------------------------------------------------------------
# STEP 17 — SEGMENT REVENUE
# ------------------------------------------------------------

if not segment_summary.empty:

    plt.figure(figsize=(8,4))

    sns.barplot(
        data=segment_summary.sort_values(
            "Revenue",
            ascending=False
        ),
        x="Customer_Segment",
        y="Revenue"
    )

    plt.title("Revenue Contribution by Customer Segment")
    plt.xlabel("Customer Segment")
    plt.ylabel("Revenue")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()


# ============================================================
# SECTION F — RFM VISUALIZATION
# ============================================================

# ------------------------------------------------------------
# STEP 18 — RFM SCORE DISTRIBUTION
# ------------------------------------------------------------

rfm_score_columns = [
    "R_Score",
    "F_Score",
    "M_Score",
    "RFM_Score"
]

available_rfm_scores = [
    col for col in rfm_score_columns
    if col in rfm.columns
]

for col in available_rfm_scores:

    plt.figure(figsize=(6,3))

    sns.countplot(
        data=rfm,
        x=col
    )

    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Number of Customers")
    plt.tight_layout()
    plt.show()


# ------------------------------------------------------------
# STEP 19 — RFM SCATTER PLOT
# ------------------------------------------------------------

if {"recency", "monetary"}.issubset(rfm.columns):

    plt.figure(figsize=(6,3))

    sns.scatterplot(
        data=rfm,
        x="recency",
        y="monetary",
        hue="Customer_Segment"
        if "Customer_Segment" in rfm.columns
        else None,
        alpha=0.6
    )

    plt.title("Recency vs Monetary Value")
    plt.xlabel("Recency")
    plt.ylabel("Monetary Value")
    plt.tight_layout()
    plt.show()


# ------------------------------------------------------------
# STEP 20 — FREQUENCY VS MONETARY
# ------------------------------------------------------------

if {"frequency", "monetary"}.issubset(rfm.columns):

    plt.figure(figsize=(6,3))

    sns.scatterplot(
        data=rfm,
        x="frequency",
        y="monetary",
        hue="Customer_Segment"
        if "Customer_Segment" in rfm.columns
        else None,
        alpha=0.6
    )

    plt.title("Frequency vs Monetary Value")
    plt.xlabel("Purchase Frequency")
    plt.ylabel("Monetary Value")
    plt.tight_layout()
    plt.show()


# ============================================================
# SECTION G — CHURN VISUALIZATION
# ============================================================

# ------------------------------------------------------------
# STEP 21 — CHURN DISTRIBUTION
# ------------------------------------------------------------

if "churn" in rfm.columns:

    churn_counts = (
        rfm["churn"]
        .value_counts()
        .sort_index()
    )

    churn_labels = [
        "Active",
        "Churned"
    ]

    plt.figure(figsize=(6,3))

    plt.pie(
        churn_counts.values,
        labels=[
            churn_labels[int(x)]
            if int(x) < len(churn_labels)
            else str(x)
            for x in churn_counts.index
        ],
        autopct="%1.1f%%",
        startangle=90
    )

    plt.title("Customer Churn Distribution")
    plt.tight_layout()
    plt.show()


# ------------------------------------------------------------
# STEP 22 — CHURN BY CUSTOMER SEGMENT
# ------------------------------------------------------------

if {
    "Customer_Segment",
    "churn"
}.issubset(rfm.columns):

    segment_churn = (
        rfm
        .groupby("Customer_Segment")
        .agg(
            Customers=("customer_id", "count"),
            Churned_Customers=("churn", "sum"),
            Churn_Rate=("churn", "mean")
        )
        .reset_index()
    )

    segment_churn["Churn_Rate"] *= 100

    plt.figure(figsize=(6,3))

    sns.barplot(
        data=segment_churn.sort_values(
            "Churn_Rate",
            ascending=False
        ),
        x="Customer_Segment",
        y="Churn_Rate"
    )

    plt.title("Churn Rate by Customer Segment")
    plt.xlabel("Customer Segment")
    plt.ylabel("Churn Rate (%)")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()


# ============================================================
# SECTION H — AGE ANALYSIS
# ============================================================

# ------------------------------------------------------------
# STEP 23 — AGE GROUP CREATION
# ------------------------------------------------------------

age_column = None

if "customer_age" in customers.columns:
    age_column = "customer_age"

elif "customer_age" in rfm.columns:
    age_column = "customer_age"

elif "Age" in customers.columns:
    age_column = "Age"


if age_column is not None:

    if "customer_age" not in customers.columns and age_column in rfm.columns:
        age_data = rfm[["customer_id", age_column]].copy()

    else:
        age_data = customers[
            ["customer_id", age_column]
        ].drop_duplicates("customer_id").copy()

    age_data["Age_Group"] = pd.cut(
        age_data[age_column],
        bins=[0, 18, 25, 35, 45, 55, 65, 100],
        labels=[
            "Under 18",
            "18-25",
            "26-35",
            "36-45",
            "46-55",
            "56-65",
            "65+"
        ],
        include_lowest=True
    )

    age_summary = (
        age_data
        .groupby("Age_Group", observed=True)
        .size()
        .reset_index(name="Customers")
    )

    plt.figure(figsize=(6,3))

    sns.barplot(
        data=age_summary,
        x="Age_Group",
        y="Customers"
    )

    plt.title("Customer Distribution by Age Group")
    plt.xlabel("Age Group")
    plt.ylabel("Number of Customers")
    plt.tight_layout()
    plt.show()


# ============================================================
# SECTION I — GENDER ANALYSIS
# ============================================================

# ------------------------------------------------------------
# STEP 24 — GENDER DISTRIBUTION
# ------------------------------------------------------------

if "gender" in customers.columns:

    gender_summary = (
        customers
        .groupby("gender")
        .agg(
            Customers=("customer_id", "nunique"),
            Revenue=("total_spend", "sum")
            if "total_spend" in customers.columns
            else ("customer_id", "count")
        )
        .reset_index()
    )

    print("\nGender Summary:")
    display(gender_summary)

    plt.figure(figsize=(6,3))

    sns.barplot(
        data=gender_summary,
        x="gender",
        y="Customers"
    )

    plt.title("Customer Distribution by Gender")
    plt.xlabel("Gender")
    plt.ylabel("Number of Customers")
    plt.tight_layout()
    plt.show()


# ============================================================
# SECTION J — PURCHASE DAY ANALYSIS
# ============================================================

# ------------------------------------------------------------
# STEP 25 — DAY OF WEEK
# ------------------------------------------------------------

transactions["Day_Name"] = transactions[
    "purchase_date"
].dt.day_name()

day_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

day_summary = (
    transactions
    .groupby("Day_Name")
    .agg(
        Transactions=("customer_id", "count"),
        Revenue=("total_purchase_amount", "sum")
    )
    .reindex(day_order)
    .reset_index()
)

# ------------------------------------------------------------
# STEP 26 — DAY OF WEEK CHART
# ------------------------------------------------------------

plt.figure(figsize=(6,3))

sns.barplot(
    data=day_summary,
    x="Day_Name",
    y="Transactions"
)

plt.title("Purchases by Day of Week")
plt.xlabel("Day")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()


# ============================================================
# SECTION K — HOURLY PURCHASE ANALYSIS
# ============================================================

# ------------------------------------------------------------
# STEP 27 — EXTRACT PURCHASE HOUR
# ------------------------------------------------------------

transactions["Purchase_Hour"] = (
    transactions["purchase_date"].dt.hour
)

hour_summary = (
    transactions
    .groupby("Purchase_Hour")
    .agg(
        Transactions=("customer_id", "count"),
        Revenue=("total_purchase_amount", "sum")
    )
    .reset_index()
)

# ------------------------------------------------------------
# STEP 28 — HOURLY PURCHASE CHART
# ------------------------------------------------------------

plt.figure(figsize=(6,3))

sns.lineplot(
    data=hour_summary,
    x="Purchase_Hour",
    y="Transactions",
    marker="o"
)

plt.title("Purchases by Hour of Day")
plt.xlabel("Hour of Day")
plt.ylabel("Number of Transactions")
plt.xticks(range(0, 24))
plt.tight_layout()
plt.show()


# ============================================================
# SECTION L — QUANTITY ANALYSIS
# ============================================================

# ------------------------------------------------------------
# STEP 29 — QUANTITY DISTRIBUTION
# ------------------------------------------------------------

quantity_summary = (
    transactions
    .groupby("quantity")
    .agg(
        Transactions=("customer_id", "count"),
        Revenue=("total_purchase_amount", "sum")
    )
    .reset_index()
)

plt.figure(figsize=(6,3))

sns.barplot(
    data=quantity_summary,
    x="quantity",
    y="Transactions"
)

plt.title("Transaction Distribution by Quantity")
plt.xlabel("Quantity Purchased")
plt.ylabel("Number of Transactions")
plt.tight_layout()
plt.show()


# ============================================================
# SECTION M — RETURN ANALYSIS
# ============================================================

# ------------------------------------------------------------
# STEP 30 — RETURN ANALYSIS
# ------------------------------------------------------------

if "returns" in transactions.columns:

    return_summary = (
        transactions
        .groupby("returns")
        .agg(
            Transactions=("customer_id", "count"),
            Revenue=("total_purchase_amount", "sum")
        )
        .reset_index()
    )

    print("\nReturn Summary:")
    display(return_summary)

    plt.figure(figsize=(6,3))

    sns.barplot(
        data=return_summary,
        x="returns",
        y="Transactions"
    )

    plt.title("Returned vs Non-Returned Transactions")
    plt.xlabel("Return Status")
    plt.ylabel("Number of Transactions")
    plt.tight_layout()
    plt.show()


# ============================================================
# SECTION N — CORRELATION HEATMAP
# ============================================================

# ------------------------------------------------------------
# STEP 31 — CUSTOMER METRIC CORRELATION
# ------------------------------------------------------------

correlation_columns = [
    "recency",
    "frequency",
    "monetary",
    "R_Score",
    "F_Score",
    "M_Score",
    "RFM_Score",
    "churn"
]

available_correlation_columns = [
    col for col in correlation_columns
    if col in rfm.columns
]

if len(available_correlation_columns) >= 2:

    correlation_matrix = rfm[
        available_correlation_columns
    ].corr()

    plt.figure(figsize=(6,3))

    sns.heatmap(
        correlation_matrix,
        annot=True,
        fmt=".2f",
        cmap="coolwarm",
        center=0
    )

    plt.title("RFM and Churn Correlation Heatmap")
    plt.tight_layout()
    plt.show()


# ============================================================
# SECTION O — TOP CUSTOMERS
# ============================================================

# ------------------------------------------------------------
# STEP 32 — TOP 10 CUSTOMERS BY REVENUE
# ------------------------------------------------------------

if "monetary" in rfm.columns:

    top_customers = (
        rfm
        .sort_values(
            "monetary",
            ascending=False
        )
        .head(10)
        .copy()
    )

    display(top_customers)

    plt.figure(figsize=(6,3))

    sns.barplot(
        data=top_customers,
        x="customer_id",
        y="monetary"
    )

    plt.title("Top 10 Customers by Monetary Value")
    plt.xlabel("Customer ID")
    plt.ylabel("Total Spend")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()


# ============================================================
# SECTION P — HIGH-VALUE AT-RISK CUSTOMERS
# ============================================================

# ------------------------------------------------------------
# STEP 33 — IDENTIFY HIGH-VALUE AT-RISK CUSTOMERS
# ------------------------------------------------------------

if {
    "R_Score",
    "M_Score",
    "churn"
}.issubset(rfm.columns):

    high_value_at_risk = rfm[
        (
            (rfm["R_Score"] <= 2)
            &
            (rfm["M_Score"] >= 4)
            &
            (rfm["churn"] == 0)
        )
    ].copy()

    high_value_at_risk = high_value_at_risk.sort_values(
        "monetary",
        ascending=False
    )

    print(
        "\nHigh-Value At-Risk Customers:",
        len(high_value_at_risk)
    )

    display(
        high_value_at_risk.head(20)
    )


# ============================================================
# SECTION Q — DASHBOARD-READY KPI TABLE
# ============================================================

# ------------------------------------------------------------
# STEP 34 — CREATE KPI TABLE
# ------------------------------------------------------------

dashboard_kpis = pd.DataFrame({
    "KPI": [
        "Total Customers",
        "Total Transactions",
        "Total Revenue",
        "Average Order Value",
        "Total Quantity Sold",
        "Total Returns",
        "Churn Rate (%)"
    ],
    "Value": [
        total_customers,
        total_transactions,
        round(total_revenue, 2),
        round(average_order_value, 2),
        total_quantity,
        total_returns,
        round(churn_rate, 2)
    ]
})

print("\n================ DASHBOARD KPI TABLE ================\n")

display(dashboard_kpis)


# ============================================================
# SECTION R — EXPORT VISUALIZATION DATA
# ============================================================

# ------------------------------------------------------------
# STEP 35 — CREATE OUTPUT FOLDER
# ------------------------------------------------------------

output_folder = Path("step6_visualization_outputs")

output_folder.mkdir(
    exist_ok=True
)


# ------------------------------------------------------------
# STEP 36 — SAVE KPI DATA
# ------------------------------------------------------------

dashboard_kpis.to_csv(
    output_folder / "dashboard_kpis.csv",
    index=False
)


# ------------------------------------------------------------
# STEP 37 — SAVE MONTHLY SALES
# ------------------------------------------------------------

monthly_sales.to_csv(
    output_folder / "monthly_sales_visualization.csv",
    index=False
)


# ------------------------------------------------------------
# STEP 38 — SAVE YEARLY SALES
# ------------------------------------------------------------

yearly_sales.to_csv(
    output_folder / "yearly_sales_visualization.csv",
    index=False
)


# ------------------------------------------------------------
# STEP 39 — SAVE CATEGORY SUMMARY
# ------------------------------------------------------------

category_summary.to_csv(
    output_folder / "category_visualization.csv",
    index=False
)


# ------------------------------------------------------------
# STEP 40 — SAVE PAYMENT SUMMARY
# ------------------------------------------------------------

payment_summary.to_csv(
    output_folder / "payment_method_visualization.csv",
    index=False
)


# ------------------------------------------------------------
# STEP 41 — SAVE DAY SUMMARY
# ------------------------------------------------------------

day_summary.to_csv(
    output_folder / "day_of_week_visualization.csv",
    index=False
)


# ------------------------------------------------------------
# STEP 42 — SAVE HOURLY SUMMARY
# ------------------------------------------------------------

hour_summary.to_csv(
    output_folder / "hourly_purchase_visualization.csv",
    index=False
)


# ------------------------------------------------------------
# STEP 43 — SAVE QUANTITY SUMMARY
# ------------------------------------------------------------

quantity_summary.to_csv(
    output_folder / "quantity_visualization.csv",
    index=False
)


# ------------------------------------------------------------
# STEP 44 — SAVE SEGMENT SUMMARY
# ------------------------------------------------------------

if not segment_summary.empty:

    segment_summary.to_csv(
        output_folder / "customer_segment_visualization.csv",
        index=False
    )


# ------------------------------------------------------------
# STEP 45 — SAVE CHURN BY SEGMENT
# ------------------------------------------------------------

if "segment_churn" in globals():

    segment_churn.to_csv(
        output_folder / "segment_churn_visualization.csv",
        index=False
    )


# ------------------------------------------------------------
# STEP 46 — SAVE AGE SUMMARY
# ------------------------------------------------------------

if "age_summary" in globals():

    age_summary.to_csv(
        output_folder / "age_group_visualization.csv",
        index=False
    )


# ------------------------------------------------------------
# STEP 47 — SAVE GENDER SUMMARY
# ------------------------------------------------------------

if "gender_summary" in globals():

    gender_summary.to_csv(
        output_folder / "gender_visualization.csv",
        index=False
    )


# ============================================================
# SECTION S — CREATE MASTER VISUALIZATION DATASET
# ============================================================

# ------------------------------------------------------------
# STEP 48 — MASTER SUMMARY
# ------------------------------------------------------------

master_visualization_summary = pd.DataFrame({
    "Metric": [
        "Total Customers",
        "Total Transactions",
        "Total Revenue",
        "Average Order Value",
        "Total Quantity",
        "Average Quantity",
        "Total Returns",
        "Churn Rate"
    ],
    "Value": [
        total_customers,
        total_transactions,
        total_revenue,
        average_order_value,
        total_quantity,
        average_quantity,
        total_returns,
        churn_rate
    ]
})

master_visualization_summary.to_csv(
    output_folder / "master_visualization_summary.csv",
    index=False
)


# ============================================================
# SECTION T — FINAL OUTPUT CHECK
# ============================================================

# ------------------------------------------------------------
# STEP 49 — LIST GENERATED FILES
# ------------------------------------------------------------

print("\n====================================================")
print("STEP 6 COMPLETED SUCCESSFULLY")
print("====================================================")

print("\nVisualization Output Files:\n")

for file in sorted(output_folder.iterdir()):
    print("✓", file.name)


# ------------------------------------------------------------
# STEP 50 — FINAL DASHBOARD KPI DISPLAY
# ------------------------------------------------------------

print("\n================ FINAL DASHBOARD KPIs ================\n")

print(f"Total Customers       : {total_customers:,}")
print(f"Total Transactions    : {total_transactions:,}")
print(f"Total Revenue         : {total_revenue:,.2f}")
print(f"Average Order Value   : {average_order_value:,.2f}")
print(f"Total Quantity Sold   : {total_quantity:,}")
print(f"Total Returns         : {total_returns:,}")
print(f"Churn Rate            : {churn_rate:.2f}%")

print("\nStep 6 visualization and dashboard preparation completed.")
```
