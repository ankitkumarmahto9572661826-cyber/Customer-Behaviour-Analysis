```python
# ============================================================
#         RFM ANALYSIS & CUSTOMER SEGMENTATION
# ============================================================

# ------------------------------------------------------------
# 1. IMPORT REQUIRED LIBRARIES
# ------------------------------------------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


# ------------------------------------------------------------
# 2. LOAD CUSTOMER-LEVEL RFM BASE DATA
# ------------------------------------------------------------

# Load the RFM base file created in Phase 2
rfm = pd.read_csv("customer_rfm_base.csv")

# Display first few records
print("First 5 records:")
display(rfm.head())

# Check dataset shape
print("Shape:", rfm.shape)

# Check column names
print("\nColumns:")
print(rfm.columns.tolist())


# ------------------------------------------------------------
# 3. BASIC DATA CHECK
# ------------------------------------------------------------

# Check data types
print("\nData Types:")
print(rfm.dtypes)

# Check missing values
print("\nMissing Values:")
print(rfm.isnull().sum())

# Check duplicate customers
print("\nDuplicate Customer IDs:", rfm["customer_id"].duplicated().sum())


# ------------------------------------------------------------
# 4. CHECK RFM STATISTICS
# ------------------------------------------------------------

# Display descriptive statistics
print("\nRFM Statistics:")
display(
    rfm[
        ["recency", "frequency", "monetary"]
    ].describe()
)


# ------------------------------------------------------------
# 5. CHECK RFM DISTRIBUTIONS
# ------------------------------------------------------------

# Create separate plots for Recency, Frequency and Monetary

plt.figure(figsize=(8, 5))
sns.histplot(rfm["recency"], bins=30, kde=True)
plt.title("Recency Distribution")
plt.xlabel("Recency (Days)")
plt.ylabel("Number of Customers")
plt.show()


plt.figure(figsize=(8, 5))
sns.histplot(rfm["frequency"], bins=30, kde=True)
plt.title("Frequency Distribution")
plt.xlabel("Number of Purchases")
plt.ylabel("Number of Customers")
plt.show()


plt.figure(figsize=(8, 5))
sns.histplot(rfm["monetary"], bins=30, kde=True)
plt.title("Monetary Distribution")
plt.xlabel("Total Spending")
plt.ylabel("Number of Customers")
plt.show()


# ------------------------------------------------------------
# 6. CREATE RFM SCORES
# ------------------------------------------------------------

# RFM scoring:
#
# Recency:
#   Lower recency = Better customer
#   Recent customers receive higher scores
#
# Frequency:
#   Higher frequency = Better customer
#
# Monetary:
#   Higher spending = Better customer


# ------------------------------------------------------------
# 6A. FUNCTION FOR QUANTILE-BASED SCORING
# ------------------------------------------------------------

def create_quantile_score(series, reverse=False):
    """
    Convert a numerical column into a score from 1 to 5
    using quantiles.

    reverse=True:
        Lower values receive higher scores.
        Used for Recency.

    reverse=False:
        Higher values receive higher scores.
        Used for Frequency and Monetary.
    """

    # Rank values first to handle large numbers of tied values
    ranked_series = series.rank(method="first")

    # Create five quantile groups
    scores = pd.qcut(
        ranked_series,
        q=5,
        labels=False
    ) + 1

    # Reverse the score for Recency
    if reverse:
        scores = 6 - scores

    return scores.astype(int)


# ------------------------------------------------------------
# 6B. CALCULATE RFM SCORES
# ------------------------------------------------------------

# Recency score
# Recent customers = high score
rfm["R_Score"] = create_quantile_score(
    rfm["recency"],
    reverse=True
)

# Frequency score
# Frequent customers = high score
rfm["F_Score"] = create_quantile_score(
    rfm["frequency"],
    reverse=False
)

# Monetary score
# High-spending customers = high score
rfm["M_Score"] = create_quantile_score(
    rfm["monetary"],
    reverse=False
)


# ------------------------------------------------------------
# 7. CREATE COMBINED RFM SCORE
# ------------------------------------------------------------

# Combine the three individual scores
rfm["RFM_Score"] = (
    rfm["R_Score"] +
    rfm["F_Score"] +
    rfm["M_Score"]
)

# Create a combined score such as 555, 453, 215 etc.
rfm["RFM_Score_Code"] = (
    rfm["R_Score"].astype(str) +
    rfm["F_Score"].astype(str) +
    rfm["M_Score"].astype(str)
)


# ------------------------------------------------------------
# 8. DISPLAY RFM SCORES
# ------------------------------------------------------------

display(
    rfm[
        [
            "customer_id",
            "recency",
            "frequency",
            "monetary",
            "R_Score",
            "F_Score",
            "M_Score",
            "RFM_Score",
            "RFM_Score_Code"
        ]
    ].head(20)
)


# ------------------------------------------------------------
# 9. CHECK SCORE DISTRIBUTIONS
# ------------------------------------------------------------

print("\nR Score Distribution:")
print(rfm["R_Score"].value_counts().sort_index())

print("\nF Score Distribution:")
print(rfm["F_Score"].value_counts().sort_index())

print("\nM Score Distribution:")
print(rfm["M_Score"].value_counts().sort_index())


# ------------------------------------------------------------
# 10. CREATE BUSINESS-ORIENTED CUSTOMER SEGMENTS
# ------------------------------------------------------------

def assign_rfm_segment(row):
    """
    Assign a business-friendly customer segment
    based on RFM scores.
    """

    R = row["R_Score"]
    F = row["F_Score"]
    M = row["M_Score"]

    # --------------------------------------------------------
    # Champions
    # --------------------------------------------------------
    if R >= 4 and F >= 4 and M >= 4:
        return "Champions"

    # --------------------------------------------------------
    # Loyal Customers
    # --------------------------------------------------------
    elif R >= 3 and F >= 4 and M >= 3:
        return "Loyal Customers"

    # --------------------------------------------------------
    # Potential Loyalists
    # --------------------------------------------------------
    elif R >= 4 and F >= 2 and M >= 2:
        return "Potential Loyalists"

    # --------------------------------------------------------
    # New Customers
    # --------------------------------------------------------
    elif R >= 4 and F <= 2:
        return "New Customers"

    # --------------------------------------------------------
    # At Risk
    # --------------------------------------------------------
    elif R <= 2 and F >= 3 and M >= 3:
        return "At Risk"

    # --------------------------------------------------------
    # Cannot Lose Them
    # --------------------------------------------------------
    elif R <= 2 and F >= 4 and M >= 4:
        return "Cannot Lose Them"

    # --------------------------------------------------------
    # Need Attention
    # --------------------------------------------------------
    elif R <= 3 and F <= 3 and M >= 3:
        return "Need Attention"

    # --------------------------------------------------------
    # Hibernating
    # --------------------------------------------------------
    elif R <= 2 and F <= 2:
        return "Hibernating"

    # --------------------------------------------------------
    # Other Customers
    # --------------------------------------------------------
    else:
        return "Others"


# Apply segmentation
rfm["Customer_Segment"] = rfm.apply(
    assign_rfm_segment,
    axis=1
)


# ------------------------------------------------------------
# 11. CHECK CUSTOMER SEGMENT COUNTS
# ------------------------------------------------------------

segment_counts = (
    rfm["Customer_Segment"]
    .value_counts()
    .reset_index()
)

segment_counts.columns = [
    "Customer_Segment",
    "Customer_Count"
]

print("\nCustomer Segment Counts:")
display(segment_counts)


# ------------------------------------------------------------
# 12. CALCULATE SEGMENT PERCENTAGE
# ------------------------------------------------------------

total_customers = len(rfm)

segment_counts["Customer_Percentage"] = (
    segment_counts["Customer_Count"]
    / total_customers
    * 100
)

print("\nCustomer Segment Distribution:")
display(segment_counts)


# ------------------------------------------------------------
# 13. CREATE SEGMENT SUMMARY
# ------------------------------------------------------------

segment_summary = (
    rfm
    .groupby("Customer_Segment")
    .agg(
        Customer_Count=("customer_id", "count"),
        Total_Revenue=("monetary", "sum"),
        Average_Spend=("monetary", "mean"),
        Average_Frequency=("frequency", "mean"),
        Average_Recency=("recency", "mean"),
        Average_RFM_Score=("RFM_Score", "mean"),
        Churn_Rate=("churn", "mean")
    )
    .reset_index()
)

# Convert churn into percentage
segment_summary["Churn_Rate"] = (
    segment_summary["Churn_Rate"] * 100
)

# Calculate customer percentage
segment_summary["Customer_Percentage"] = (
    segment_summary["Customer_Count"]
    / total_customers
    * 100
)

# Round numerical columns
segment_summary = segment_summary.round(2)

# Sort by total revenue
segment_summary = segment_summary.sort_values(
    "Total_Revenue",
    ascending=False
)

print("\nRFM Segment Summary:")
display(segment_summary)


# ------------------------------------------------------------
# 14. VISUALIZE CUSTOMER SEGMENT DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

segment_plot = (
    rfm["Customer_Segment"]
    .value_counts()
    .sort_values(ascending=False)
)

sns.barplot(
    x=segment_plot.values,
    y=segment_plot.index
)

plt.title("Customer Segment Distribution")
plt.xlabel("Number of Customers")
plt.ylabel("Customer Segment")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 15. VISUALIZE REVENUE BY CUSTOMER SEGMENT
# ------------------------------------------------------------

revenue_plot = (
    rfm
    .groupby("Customer_Segment")["monetary"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=revenue_plot.values,
    y=revenue_plot.index
)

plt.title("Total Revenue by Customer Segment")
plt.xlabel("Total Revenue")
plt.ylabel("Customer Segment")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 16. VISUALIZE AVERAGE CUSTOMER SPENDING
# ------------------------------------------------------------

avg_spend_plot = (
    rfm
    .groupby("Customer_Segment")["monetary"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=avg_spend_plot.values,
    y=avg_spend_plot.index
)

plt.title("Average Customer Spending by Segment")
plt.xlabel("Average Spending")
plt.ylabel("Customer Segment")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 17. VISUALIZE AVERAGE RECENCY
# ------------------------------------------------------------

avg_recency_plot = (
    rfm
    .groupby("Customer_Segment")["recency"]
    .mean()
    .sort_values(ascending=True)
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=avg_recency_plot.values,
    y=avg_recency_plot.index
)

plt.title("Average Recency by Customer Segment")
plt.xlabel("Average Recency (Days)")
plt.ylabel("Customer Segment")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 18. VISUALIZE AVERAGE PURCHASE FREQUENCY
# ------------------------------------------------------------

avg_frequency_plot = (
    rfm
    .groupby("Customer_Segment")["frequency"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=avg_frequency_plot.values,
    y=avg_frequency_plot.index
)

plt.title("Average Purchase Frequency by Segment")
plt.xlabel("Average Purchase Frequency")
plt.ylabel("Customer Segment")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 19. RFM SCORE HEATMAP
# ------------------------------------------------------------

rfm_heatmap = (
    rfm
    .groupby(
        ["R_Score", "F_Score"]
    )["M_Score"]
    .mean()
    .unstack()
)

plt.figure(figsize=(8, 6))

sns.heatmap(
    rfm_heatmap,
    annot=True,
    fmt=".2f"
)

plt.title("Average Monetary Score by Recency and Frequency")
plt.xlabel("Frequency Score")
plt.ylabel("Recency Score")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 20. ANALYZE CHURN BY RFM SEGMENT
# ------------------------------------------------------------

churn_by_segment = (
    rfm
    .groupby("Customer_Segment")
    .agg(
        Customers=("customer_id", "count"),
        Churned_Customers=("churn", "sum"),
        Churn_Rate=("churn", "mean")
    )
    .reset_index()
)

# Convert churn rate to percentage
churn_by_segment["Churn_Rate"] = (
    churn_by_segment["Churn_Rate"] * 100
)

churn_by_segment = churn_by_segment.round(2)

# Sort by churn rate
churn_by_segment = churn_by_segment.sort_values(
    "Churn_Rate",
    ascending=False
)

print("\nChurn by Customer Segment:")
display(churn_by_segment)


# ------------------------------------------------------------
# 21. VISUALIZE CHURN RATE BY SEGMENT
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

sns.barplot(
    data=churn_by_segment,
    x="Churn_Rate",
    y="Customer_Segment"
)

plt.title("Churn Rate by Customer Segment")
plt.xlabel("Churn Rate (%)")
plt.ylabel("Customer Segment")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 22. IDENTIFY HIGH-VALUE CUSTOMERS
# ------------------------------------------------------------

# Customers with high RFM scores
high_value_customers = rfm[
    rfm["RFM_Score"] >= 12
].copy()

print(
    "Number of High-Value Customers:",
    len(high_value_customers)
)

display(
    high_value_customers[
        [
            "customer_id",
            "customer_name",
            "recency",
            "frequency",
            "monetary",
            "RFM_Score",
            "Customer_Segment",
            "churn"
        ]
    ]
    .sort_values("RFM_Score", ascending=False)
    .head(20)
)


# ------------------------------------------------------------
# 23. IDENTIFY AT-RISK CUSTOMERS
# ------------------------------------------------------------

at_risk_customers = rfm[
    rfm["Customer_Segment"].isin(
        ["At Risk", "Cannot Lose Them"]
    )
].copy()

print(
    "Number of At-Risk / Cannot-Lose Customers:",
    len(at_risk_customers)
)

display(
    at_risk_customers[
        [
            "customer_id",
            "customer_name",
            "recency",
            "frequency",
            "monetary",
            "RFM_Score",
            "Customer_Segment",
            "churn"
        ]
    ]
    .sort_values("monetary", ascending=False)
    .head(20)
)


# ------------------------------------------------------------
# 24. IDENTIFY HIGH-VALUE CHURNED CUSTOMERS
# ------------------------------------------------------------

high_value_churned = rfm[
    (rfm["churn"] == 1) &
    (rfm["RFM_Score"] >= 10)
].copy()

print(
    "High-Value Churned Customers:",
    len(high_value_churned)
)

display(
    high_value_churned[
        [
            "customer_id",
            "customer_name",
            "recency",
            "frequency",
            "monetary",
            "RFM_Score",
            "Customer_Segment"
        ]
    ]
    .sort_values("monetary", ascending=False)
    .head(20)
)


# ------------------------------------------------------------
# 25. SAVE COMPLETE RFM SEGMENTATION DATASET
# ------------------------------------------------------------

rfm.to_csv(
    "customer_rfm_segments.csv",
    index=False
)

print(
    "Saved: customer_rfm_segments.csv"
)


# ------------------------------------------------------------
# 26. SAVE SEGMENT SUMMARY
# ------------------------------------------------------------

segment_summary.to_csv(
    "rfm_segment_summary.csv",
    index=False
)

print(
    "Saved: rfm_segment_summary.csv"
)


# ------------------------------------------------------------
# 27. SAVE HIGH-VALUE CUSTOMERS
# ------------------------------------------------------------

high_value_customers.to_csv(
    "high_value_customers.csv",
    index=False
)

print(
    "Saved: high_value_customers.csv"
)


# ------------------------------------------------------------
# 28. SAVE AT-RISK CUSTOMERS
# ------------------------------------------------------------

at_risk_customers.to_csv(
    "at_risk_customers.csv",
    index=False
)

print(
    "Saved: at_risk_customers.csv"
)


# ------------------------------------------------------------
# 29. SAVE HIGH-VALUE CHURNED CUSTOMERS
# ------------------------------------------------------------

high_value_churned.to_csv(
    "high_value_churned_customers.csv",
    index=False
)

print(
    "Saved: high_value_churned_customers.csv"
)


# ------------------------------------------------------------
# 30. FINAL STEP 3 CHECK
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("PHASE 3 COMPLETED")
print("=" * 60)

print("\nTotal Customers:", len(rfm))

print(
    "Average Recency:",
    round(rfm["recency"].mean(), 2),
    "days"
)

print(
    "Average Frequency:",
    round(rfm["frequency"].mean(), 2)
)

print(
    "Average Monetary Value:",
    round(rfm["monetary"].mean(), 2)
)

print(
    "Average RFM Score:",
    round(rfm["RFM_Score"].mean(), 2)
)

print("\nCustomer Segments:")
print(
    rfm["Customer_Segment"]
    .value_counts()
)

print("\nOutput files created:")
print("1. customer_rfm_segments.csv")
print("2. rfm_segment_summary.csv")
print("3. high_value_customers.csv")
print("4. at_risk_customers.csv")
print("5. high_value_churned_customers.csv")
```
